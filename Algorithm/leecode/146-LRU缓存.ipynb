### 146.LRU缓存
运用你所掌握的数据结构，设计和实现一个 LRU (最近最少使用) 缓存机制。它应该支持以下操作：获取数据 `get` 和写入数据 `put` 。
get(key) - 如果密钥 (key) 存在于缓存中，则获取密钥的值（总是正数），否则返回 -1。
put(key, value) - 如果密钥不存在，则写入其数据值。当缓存容量达到上限时，它应该在写入新数据之前删除最近最少使用的数据值，从而为新的数据值留出空间。
进阶:
你是否可以在 O(1) 时间复杂度内完成这两种操作？

```python
class DListNode:
    def __init__(self, key=0, value=0):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None


class LRUCache:
    # 双向链表：维护访问顺序
    # 哈希表（dict），存储并快速找到节点

    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = {} # dict key->node
        self.head = DListNode()
        self.tail = DListNode()
        # 初始化链表尾头尾相连
        self.head.next = self.tail
        self.tail.prev = self.head
        

    def get(self, key: int) -> int: # 哈希表中查找key
        # 不存在返回-1
        if key not in self.cache:
            return -1

        #存在，通过哈希表找到对应链表节点，并移动到链表头
        node = self.cache[key]
        self._move2head(node)
        return node.value
        

    def put(self, key: int, value: int) -> None: # 哈希表中找key
        # 存在，更新value
        if key in self.cache:
            node = self.cache[key]
            node.value = value
            self._move2head(node)
        else:
        # 不存在，新建插入链表头
            node = DListNode(key, value)
            self.cache[key] = node # 存储的是node的位置指针
            self._add2head(node)

        # 判断容量大小，超过则删除链表尾部节点
        if len(self.cache) > self.capacity:
            remove = self._remove_tail()
            del self.cache[remove.key]
        

    def _add2head(self, node):
        """将节点插入head"""
        node.next = self.head.next # self.head 是 dummy_head
        self.head.next.prev = node
        node.prev = self.head
        self.head.next = node


    def _move2head(self, node):
        """将节点移到head"""
        self._remove(node)
        self._add2head(node)


    def _remove(self, node):
        """删除节点"""
        node.prev.next = node.next
        node.next.prev = node.prev
    
    def _remove_tail(self) -> node:
        """删除尾节点并返回"""
        node = self.tail.prev # self.tail 是 dummy_tail
        self._remove(node)
        return node # 返回删除的节点
    
# Your LRUCache object will be instantiated and called as such:
# obj = LRUCache(capacity)
# param_1 = obj.get(key)
# obj.put(key,value)
```
