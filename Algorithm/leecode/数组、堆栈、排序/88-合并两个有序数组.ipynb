```python
# 使用了额外空间
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        # 双指针法
        tmp = []
        left, right = 0, 0
        while left < m and right < n: # 两指针都未越界
            if nums1[left] < nums2[right]:
                tmp.append(nums1[left])
                left += 1
            else:
                tmp.append(nums2[right])
                right += 1

        while left < m: # nums1还有剩余
            tmp.append(nums1[left])
            left += 1
        while right < n:
            tmp.append(nums2[right])
            right += 1

        nums1[0: m + n] = tmp # 索引切片左闭右开，不包含最后
        return nums1
```

```python
# 不使用额外空间，双指针法，逆向双指针
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        # 双指针法，逆向双指针
        last = m + n - 1
        left, right = m - 1, n - 1

        # nums1, nums2本身有序，遍历排序完nums2, 在原本nums1基础上本就有序
        while right >= 0: # 从后往前遍历
            if left >= 0 and nums1[left] > nums2[right]:
                nums1[last] = nums1[left]
                left -= 1
            else:
                nums1[last] = nums2[right]
                right -= 1
            last -= 1 # 数组每轮必从后向前加入值
```
