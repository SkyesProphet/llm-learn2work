### 153. 寻找旋转排序数组的最小值
**关联题目：33-搜索旋转排序数组**
已知一个长度为 n 的数组，预先按照升序排列，经由 1 到 n 次 旋转 后，得到输入数组。例如，原数组 nums = [0,1,2,4,5,6,7] 在变化后可能得到：
若旋转 4 次，则可以得到 [4,5,6,7,0,1,2]
若旋转 7 次，则可以得到 [0,1,2,4,5,6,7]
注意，数组 [a[0], a[1], a[2], ..., a[n-1]] 旋转一次 的结果为数组 [a[n-1], a[0], a[1], a[2], ..., a[n-2]] 。

你必须设计一个时间复杂度为 O(log n) 的算法解决此问题。

二分的范围可以是 (−1,n−1)，也就是闭区间 [0,n−2]。

这是因为，如果 nums[n−1] 是数组最小值，那么 nums 分成两段，第一段 [0,n−2]，第二段 [n−1,n−1]，且第一段的所有数都大于 nums[n−1]。每次 x 和 nums[n−1] 比大小，一定是 x>nums[n−1]。这意味着每次二分更新的都是 left，那么循环结束后，答案自然就是 n−1 了。

作者：灵茶山艾府
链接：https://leetcode.cn/problems/find-minimum-in-rotated-sorted-array/

```python
from typing import List
class Solution:
    def findMin(self, nums: List[int]) -> int:
        left, right = 0, len(nums) - 1  # 闭区间 [0, n-1]

        #旋转点为原序列 0下标，即为最小值
        while left < right: # 开区间不为空
            mid = (left + right) // 2

            if nums[mid] < nums[-1]: # 旋转点在 左
                right = mid
            else:
                left = mid + 1
        return nums[left]
```

```python
class Solution:
    def findMin(self, nums: List[int]) -> int:
        left, right = -1, len(nums) - 1  # 开区间 (-1, n-1)
        while left + 1 < right:  # 开区间不为空
            mid = (left + right) // 2
            if nums[mid] < nums[-1]:
                right = mid
            else:
                left = mid
        return nums[right]
```
