### 912.排序数组 
给你一个整数数组 nums，请你将该数组升序排列。

你必须在 不使用任何内置函数 的情况下解决问题，时间复杂度为 O(nlog(n))，并且空间复杂度尽可能小。

输入：nums = [5,2,3,1]
输出：[1,2,3,5]
解释：数组排序后，某些数字的位置没有改变（例如，2 和 3），而其他数字的位置发生了改变（例如，1 和 5）。

#### 1. 快速排序（分区交换排序）
    - 在平均状况下，排序 n 个项目要 `O(nlogn)` 次比较。在最坏状况下则需要 `O(n^2)` 次比较，但这种状况并不常见。

快速排序的核心思想：(概念解说：灵茶山艾府)

1. 定义一个递归函数 `quick_sort(left,right)`，表示把子数组 `[left_arr,right_arr]` 升序排序。
    - **挑选基准值**：随机挑选基准pivot。
    - **分割**：所有比基准值小的元素摆放在基准前面，所有比基准值大的元素摆在基准后面（与基准值相等的数可以到任何一边）
2. 划分子数组 `[left_arr, right_arr]`。设划分后，此时基准元素 pivot 的下标为 mid。
    - 在 pivot 左侧的元素，继续递归排序，即递归调用 `sort(left, mid − 1)`。
    - 在 pivot 右侧的元素，继续递归排序，即递归调用 `sort(mid + 1, right)`。
3. 递归边界：如果 `left≥right`，说明子数组为空或者只剩一个元素，此时无需排序，直接返回。
4. 递归入口：`sort(0,n−1)`。

```python
import random
from typing import List
class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        
        def quick_sort(arr, low, high):
            if(low >= high): # 子数组异常，递归结束
                return

            # 随机选择锚点
            pivot_idx = random.randint(low, high)
            # 将锚点丢到最左侧，便于遍历
            arr[low], arr[pivot_idx] = arr[pivot_idx], arr[low]
            pivot = arr[low] # 记录锚点

            left, right = low, high # 双指针
            while left < right: # 左右指针重合时，跳出
                # 从右找小于povit的
                while left < right and arr[right] > pivot:
                    right -= 1 # 继续找
                if left < right: # 找到
                    arr[left] = arr[right] # 丢到 左子数组
                    left += 1 # 找到一个后立即移动，确保left 和 right都会更新并靠近（更偏向等二分）
                
                # 从左找大于povit的
                while left < right and arr[left] < pivot:
                    left += 1 #继续找

                if left < right: # 找到
                    arr[right] = arr[left] # 丢到 右子数组
                    right -= 1 

            arr[left] = pivot # 跳出遍历，最后的位置就是povit

            mid = left # 此时mid == left == right
            # 递归子数组
            quick_sort(arr, low, mid - 1)
            quick_sort(arr, mid + 1, high)

        quick_sort(nums, 0, len(nums) -1) # 调用快速排序
        return nums
```

#### 2.归并排序（Merge Sort）
归并排序是建立在归并操作上的一种有效，稳定的排序算法，该算法是采用分治法（Divide and Conquer）的一个非常典型的应用。
分治策略：将已有序的子序列合并，得到完全有序的序列。
- 「分」的阶段首先将序列一步步分解成小的子序列进行分段排序；
- 「治」的阶段则将分段有序的子序列合并在一起，使得整个序列变得有序。
- 注：两个自子序列合并时，相对有序，合并时，左右子序列分别通过两个指针 left,right 指向，比较两个指针所指向的元素大小，将较小的元素加入到新数组中，并移动指针。

```python
from typing import List
class Solution:
    def sortArray(self, nums:List[int]) -> list[int]:

        def merge_sort(arr, low, high):
            if low < high:
                return # 子数组异常，递归结束
            
            mid = low + (high - low) // 2 # 二分子数组, 中间位置
            merge_sort(arr, low, mid) # 递归左子数组
            merge_sort(arr, mid + 1, high) # 递归右子数组

            left, right = low, mid + 1 # 双指针，分别指向左右子数组(已有序)的起始位置
            tmp = [] # 临时数组，存储合并结果
            
            while left <= mid and right <= high: # 双指针都未越界
                if arr[left] < arr[right]: # 左子数组的当前元素较小
                    tmp.append(arr[left]) # 加入临时数组
                    left += 1 # 左指针右移
                else: # 右子数组的当前元素较小
                    tmp.append(arr[right]) # 加入临时数组
                    right += 1 # 右指针右移

            while left <= mid: # 左子数组还有剩余元素
                tmp.append(arr[left]) # 加入临时数组
                left += 1 # 左指针右移

            while right <= high: # 右子数组还有剩余元素
                tmp.append(arr[right]) # 加入临时数组
                right += 1 # 右指针右移
            
            arr[low:high+1] = tmp
        
        merge_sort(nums, 0, len(nums) - 1) # 调用归并排序
        return nums
```

#### 3. 堆排序（Heap Sort）
指利用堆（heap）这种数据结构所设计的一种排序算法。堆是一个近似完全二叉树的结构，并同时满足堆的性质：即子节点的键值或索引总是小于（或者大于）它的父节点。
1. 堆可看作是一个「完全二叉树」的结构：
   - 大根堆（大顶堆）：每个节点的值均大于等于其左右孩子节点的值；`nums[i]>=nums[2i+1] & nums[i]>=nums[2i+2]` 从 0 开始索引
   - 小根堆（小顶堆）：每个节点的值均小于等于其左右孩子节点的值。`nums[i]<=nums[2i+1] & nums[i]<=nums[2i+2]`

```python
class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        
        # 大顶堆，序号 0 开始
        def max_heapify(heap, root, heap_size):
            """
            heap: 待排序的堆
            root: 当前节点为根（考虑以其为根的子树）
            heap_size: 堆的大小
            """
            largest = root # 先假设当前节点最大
            
            while True: # 循环，直到当前节点为叶子节点
                left = 2 * root + 1 # 左子节点
                right = 2 * root + 2 # 右子节点

                if left < heap_size and heap[left] > heap[largest]: # 左子节点较大
                    largest = left
                if right < heap_size and heap[right] > heap[largest]: # 右子节点较大
                    largest = right
                
                if largest == root: # 当前节点最大，无需调整
                    break

                # largest 记录当前子树中最大节点的索引
                heap[root], heap[largest] = heap[largest], heap[root] # 交换当前节点和较大的子节点
                root = largest # 更新当前根节点为较大的子节点，继续调整

        n = len(nums) # 堆的大小
        # 构建大顶堆，从最后一个非叶子节点开始调整
        for i in range(n // 2 - 1, -1, -1): # 从 n // 2 - 1 to 0
            max_heapify(nums, i, n)
        
        # 进行堆排序, 每次将堆顶元素（最大）与末尾元素交换，然后调整剩余的堆
        for i in range(n - 1, 0, -1): # 从 n - 1 to 1
            nums[0], nums[i] = nums[i], nums[0] # 将堆顶元素与末尾元素交换
            max_heapify(nums, 0, i) # 重新调整堆

        return nums
```
