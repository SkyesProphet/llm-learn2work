### 33.搜索旋转排序数组
**关联题目153.寻找旋转排序数组的最小元素**
整数数组 nums 按升序排列，数组中的值 互不相同 。

在传递给函数之前，nums 在预先未知的某个下标 k（0 <= k < nums.length）上进行了 向左旋转，使数组变为 `[nums[k], nums[k+1], ..., nums[n-1], nums[0], nums[1], ..., nums[k-1]]`（下标 从 0 开始 计数）。例如， `[0,1,2,4,5,6,7]` 下标 3 上向左旋转后可能变为 `[4,5,6,7,0,1,2]` 。

给你 旋转后 的数组 nums 和一个整数 target ，如果 nums 中存在这个目标值 target ，则返回它的下标，否则返回 -1 。

你必须设计一个时间复杂度为 O(log n) 的算法解决此问题。

#### 1. 最暴力，内置函数index

```python
from typing import List
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        try: 
            return nums.index(target)
        except ValueError:
            return -1
```

#### 2.(一次)二分法，虽然数组被旋转，但至少一半有序。
1. 由于原数组升序，故旋转点左右升序。可比较首尾值，判断旋转点在哪一半。
2. 使用二分法，将数组分为两半，比较中间值与首尾值，并判断旋转点在哪一半。
3. 旋转点在无序的一边。
   1. 若左子序列有序，判断target是否在左半。
   2. 若右子序列有序，判断target是否在右半。
   3. 根据判断缩小范围。
4. 时间复杂度O(logn)，空间复杂度 O(1)

```python
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        # try: 
        #     return nums.index(target)
        # except ValueError:
        #     return -1

        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right) // 2
            
            if nums[mid] == target:
                return mid # 最终遍历的target一定是mid,退出循环

            if nums[left] <= nums[mid]: # 左子序有序，旋转点在右
                # 判断target是否在左子序列
                if nums[left] <= target < nums[mid]:
                    right = mid - 1 # 缩小范围
                else:
                    left = mid + 1
            else: # 左无序，旋转点在左，右子序列有序
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1

        return -1
```

#### （二次）二分法
**关联：153. 寻找旋转排序数组的最小值**
1. 首先找到旋转点（最小值）下标 i，然后根据旋转点将数组分为两个有序数组，分别进行二分查找。
2. 分类讨论：
   - 如果 `target>nums[n−1]`，那么 target 一定在第一段 `[0,i−1]` 中，在 `[0,i−1]` 中二分查找 target。
   - 如果 `target≤nums[n−1]`:
     - 如果 `i=0`，说明 nums 是递增的，直接在 `[0,n−1] `中二分查找 target。
    - 如果 `i>0`，那么 target 一定在第二段 `[i,n−1]` 中，在 `[i,n−1]` 中二分查找 target。
    - 这两种情况可以合并成：在 `[i,n−1]` 中二分查找 target。

```python
class Solution:
    # 153. 寻找旋转排序数组中的最小值（返回的是下标）
    def findMin(self, nums: List[int]) -> int:
        left, right = -1, len(nums) - 1  # 开区间 (-1, n-1)
        while left + 1 < right:  # 开区间不为空
            mid = (left + right) // 2
            if nums[mid] < nums[-1]:
                right = mid
            else:
                left = mid
        return right

    def search(self, nums: List[int], target: int) -> int:
        
        min_index = self.findMin(nums) # 先找到旋转点
        # min_index 之前的部分是有序的，之后的部分也是有序的
        # 判断target在哪个部分
        if nums[min_index] <= target <= nums[-1]: # 在右边有序部分
            left, right = min_index, len(nums) - 1
        else: # 在左边有序部分
            left, right = 0, min_index - 1
        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
        return -1
```
