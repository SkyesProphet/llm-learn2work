### 49. 字母异位词分组——哈希
给你一个字符串数组，请你将 字母异位词 组合在一起。可以按任意顺序返回结果列表。
 

示例 1:

输入: strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

输出: [["bat"],["nat","tan"],["ate","eat","tea"]]

解释：
- 在 strs 中没有字符串可以通过重新排列来形成 "bat"。
- 字符串 "nat" 和 "tan" 是字母异位词，因为它们可以重新排列以形成彼此。
- 字符串 "ate" ，"eat" 和 "tea" 是字母异位词，因为它们可以重新排列以形成彼此。

示例 2:

输入: strs = [""]
输出: [[""]]
解释: "" 的字母异位词之一就是自身。

示例 3:

输入: strs = ["a"]
输出: [["a"]]

提示：
- 1 <= strs.length <= 104
- 0 <= strs[i].length <= 100
- strs[i] 仅包含小写字母

#### 1. 哈希表分组
如果两个字符串从小到大排序后相等，那么两个字符串就互为字母异位词，否则不是。

例如 aab,aba,baa 排序后都是 aab，所以 aab,aba,baa 互为字母异位词。

把排序后相同的字符串分到同一组。

**算法**：
1. 用哈希表分组，把排序后的字符串当作哈希表的 key，排序前的字符串加到对应的列表中（哈希表的 value）。
2. 把哈希表的所有 value 加到一个列表中返回。


```python
from typing import List
from collections import defaultdict
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        # 如果 key 不在字典中，则自动插入一个空列表
        d = defaultdict(list)
        for s in strs:
            st = ''.join(sorted(s)) # 把 s 排序，作为哈希表的 key
            d[st].append(s) # 将原字符加入对应key
        
        return list(d.values()) # 哈希表的 value 保存分组后的结果
    
# strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
# strs = [""]
strs = ["a"]
print(Solution().groupAnagrams(strs))
```
