### 128.最长连续序列——哈希
给定一个未排序的整数数组 nums ，找出数字连续的最长序列（不要求序列元素在原数组中连续）的长度。

请你设计并实现时间复杂度为 O(n) 的算法解决此问题。


示例 1：

输入：nums = [100,4,200,1,3,2]
输出：4

解释：最长数字连续序列是 [1, 2, 3, 4]。它的长度为 4。

示例 2：

输入：nums = [0,3,7,2,5,8,4,6,0,1]
输出：9

提示：
- `0 <= nums.length <= 10^5`
- `-10^9 <= nums[i] <= 10^9`

#### 哈希表 O(n) 做法
限制条件：时间复杂度O(n)。
-  本题不能排序的，因为排序的时间复杂度是 O(nlogn)，不符合题目 O(n) 的要求。

**核心思路**：
对于 nums 中的元素 x，以 x 为起点，不断查找下一个数 x+1,x+2,⋯ 是否在 nums 中，并统计序列的长度。
时间复杂度优化：
1. 把 nums 中的数都放入一个哈希集合中，这样可以 O(1) 判断数字是否在 nums 中。
2. 如果 x−1 在哈希集合中，则不以 x 为起点。
   1. 因为以 x−1 为起点计算出的序列长度，一定比以 x 为起点计算出的序列长度要长！
   2. 以 x -1 为起点，必定比 x 为起点要长 1
3. 遍历元素的时候，要遍历哈希集合，而不是 nums。相同的元素只记录一次。

```python
from typing import List
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        st = set(nums) # hash，一个数字只出现一次
        ans = 0
        for x in st: # 枚举所有数字
            if x - 1 in st: # 如果 x-1 在哈希表中，说明 x 不是一个连续序列的起点
                continue
            
            # x 是序列起点
            y = x + 1
            while y in st: # 不断地往后找，直到找不到为止
                y += 1
            
            # 循环结束后，y-1 是最后一个在哈希集合中的数 y-1-x+1
            ans = max(ans, y - x) # 连续字符串，长度直接减
        return ans

# nums = [100,4,200,1,3,2]
nums = [0,3,7,2,5,8,4,6,0,1]
print(Solution().longestConsecutive(nums))
```
