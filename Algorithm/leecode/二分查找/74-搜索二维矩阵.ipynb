###  74.搜索二维矩阵
给你一个满足下述两条属性的 m x n 整数矩阵：

- 每行中的整数从左到右按非严格递增顺序排列。
- 每行的第一个整数大于前一行的最后一个整数。

给你一个整数 target ，如果 target 在矩阵中，返回 true ；否则，返回 false 。

示例 1：
- 输入：matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 3
- 输出：true

示例 2：
- 输入：matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 13
- 输出：false

提示：
- m == matrix.length
- n == matrix[i].length
- 1 <= m, n <= 100
- -10^4 <= matrix[i][j], target <= 10^4


#### 1. 二分查找
**分析**：矩阵的每一行是递增的，且每行的第一个数大于前一行的最后一个数，如果把矩阵每一行拼在一起，我们可以得到一个递增数组。

对于有序数组，可以用二分查找判断 target 是否在 matrix 中。
实现：
1. 代码实现时，并不需要真的拼成一个长为 mn 的数组 a，而是将 a[i] 转换成矩阵中的行号和列号。
2. 一般地，有 ` a[i] = matrix[i // n][i % n] `

```python
# 开区间二分法
from typing import List
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        m, n = len(matrix), len(matrix[0])
        left, right = -1, m * n # 开区间(-1, m*n)

        while left + 1 < right: # 区间内不为空
            mid = left + (right - left) // 2
            x = matrix[mid // n][mid % n] # 一维索引转二维索引
            if x == target:
                return True
            elif x < target:
                left = mid # 目标在右侧，更新左边界
            else:
                right = mid # 目标在左侧，更新右边界
        return False # 没有找到
    
matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]]
target = 3
print(Solution().searchMatrix(matrix, target))
```

#### 2. 排除法
分析：矩阵每行的元素递增，且每行第一个元素大于上一行最后一个元素。即每列也是递增的。可以从数列 右上角开始收缩矩阵。

步骤：
1. 从右上角 `NUM[0][n-1]` 开始，向左下角收缩。
2. 若 `nums[i][n-1] < target`，则 这一行前面的数 全小于 target。`i ++`。继续搜索下一行右上角。
3. 若 nums[i][n-1] > target，说明最后一列所有数都 大于 target。`n --`。继续搜索上一列 右上角。
4. 直到右上角元素 等于 target。结束。
5. 没找到返回：False

```python
from typing import List
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        m, n = len(matrix), len(matrix[0])
        i, j = 0, n - 1 # 从右上角开始搜索

        while i < m and j >= 0:
            if matrix[i][j] == target:
                return True
            elif matrix[i][j] < target:
                i += 1 # 这一行剩余元素全部小于 target，排除
            else:
                j -= 1 # 这一列剩余元素全部大于 target，排除
        return False
    
matrix = [
  [1,   4,  7, 11, 15],
  [2,   5,  8, 12, 19],
  [3,   6,  9, 16, 22],
  [10, 13, 14, 17, 24],
  [18, 21, 23, 26, 30]
]
target = 5
print(Solution().searchMatrix(matrix, target))
```
