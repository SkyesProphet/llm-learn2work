### 34. 在排序数组中查找元素的第一个和最后一个位置
给你一个按照非递减顺序排列的整数数组 nums，和一个目标值 target。请你找出给定目标值在数组中的开始位置和结束位置。

如果数组中不存在目标值 target，返回 [-1, -1]。

你必须设计并实现**时间复杂度为 O(log n) 的算法**解决此问题。

示例 1：

输入：nums = [5,7,7,8,8,10], target = 8
输出：[3,4]

示例 2：

输入：nums = [5,7,7,8,8,10], target = 6
输出：[-1,-1]

示例 3：

输入：nums = [], target = 0
输出：[-1,-1]


#### 1.二分查找
有三种写法：闭区间、左闭右开、开区间.

相关题目：**35.搜索插入位置**。 实现方法一样。
- 实现函数 `lower_bound(nums, target)` ,找到 `nums[i] >= target` 的的最左边位置。

- 找到 target 的起始位置，即找到 `num[i] >= target` 的第一个位置,最左边。二分法
- 找到 target 的结束位置，即找到 `num[i] > target` 的第一个位置。

以 `num[i] >= target` 为例可对 `>= , > , < , <=` 进行处理。
1. 求 `num[i] > target` 的最左边位置， 即找 `numm[i] >= target + 1 > target` 的第一个位置，最左位置。
2. 求 `num[i] < target` 的最右边位置， 即找 `numm[i] >= target` 的最左位置 -1。
3. 求 `num[i] <= target` 的最右边位置， 即找 `numm[i] > target` 的最左位置 -1。

求 target 的开始位置和结束位置 = 求 `num[i] >= target` 的最左位置 和 `num[j] <= target` 的最右位置。
1. 设计函数 lower_bound(nums, target)，求出 `num[i] >= target` 的最左位置。
   - `start = lower_bound(nums, target)`
2. 根据 `num[i] <= target` 的最右边位置，等价于 `num[j] > target` 的最左位置 -1。即 `num[j] >= target + 1` 的最左位置 -1。
   - 在所有数都是整数的前提下，>target 等价于 ≥target+1，这样就可以复用我们已经写好的二分函数了。
   - `end = lower_bound(nums, target + 1) - 1`

**三种区间**写法如何区分：
1. 怎么判断我写的是哪一种二分？
- 答：看 while 循环的条件，如果是 left <= right，就是闭区间；如果是 left < right，就是半闭半开区间；如果是 left + 1 < right，就是开区间。
2. 对于闭区间写法，为什么 nums[mid] >= target 的时候要写 right = mid - 1？此时的 mid 不是有可能是答案吗？这样写不会错过答案吗？
- 答：即根据 **答案在区间外面，不在区间里面**，闭区间循环结束后 left>right，区间 [left,right] 是空的，什么也没有。

```python
# 左闭右开
from typing import List 
class Solution:
    # lower_bound 返回最小的满足 nums[i] >= target 的下标 i (左起始)
    def lower_bound(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums)  # 左闭右开 [left, right)

        while left < right: # 此时区间不为空
            # 循环不变量
            # nums[left -1] < target
            # nums[right] >= target
            mid = left + (right - left) // 2
            if nums[mid] < target:
                left = mid + 1 # [mid +1, right)
            else:
                right = mid  # [left,mid)
        return left 

    def searchRange(self, nums: List[int], target: int) -> List[int]:
        # 起始位置，找到 nums[start] >= target 的最小下标 start
        start = self.lower_bound(nums, target)

        # 先判断 start 是否越界，或者 nums[start] != target,说明不存在 target
        if start == len(nums) or nums[start] != target: 
            return [-1, -1]
        # 若start 存在，必存在 end, start <= end
        end = self.lower_bound(nums, target + 1) - 1
        return [start, end]
    
# nums = [5,7,7,8,8,10]
# target = 8
# nums = [5,7,7,8,8,10]
# target = 6
nums = []
target = 0
print(Solution().searchRange(nums, target))
```

```python
# 开区间（开区间更好记）
class Solution:
    # lower_bound 返回最小的满足 nums[i] >= target 的下标 i (左起始)
    def lower_bound(self, nums: List[int], target: int) -> int:
        left, right = -1, len(nums)  # 开区间 (left, right)

        while left + 1 < right: # 此时区间不为空
            # 循环不变量
            # nums[left] < target
            # nums[right] >= target
            mid = left + (right - left) // 2
            if nums[mid] < target:
                left = mid  # (mid, right)
            else:
                right = mid  # (left, mid)

            # 循环结束后 left+1 = right
        # 此时 nums[left] < target 而 nums[right] >= target
        # 所以 right 就是第一个 >= target 的元素下标
        return right # right = left + 1
    

    def searchRange(self, nums: List[int], target: int) -> List[int]:
        # 起始位置，找到 nums[start] >= target 的最小下标 start
        start = self.lower_bound(nums, target)

        # 先判断 start 是否越界，或者 nums[start] != target,说明不存在 target
        if start == len(nums) or nums[start] != target: 
            return [-1, -1]
        # 若start 存在，必存在 end, start <= end
        end = self.lower_bound(nums, target + 1) - 1
        return [start, end]
```
