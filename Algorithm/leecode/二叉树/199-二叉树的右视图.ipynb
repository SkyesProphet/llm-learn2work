### 199. 二叉树的右视图
给定一个二叉树的 根节点 root，想象自己站在它的右侧，按照从顶部到底部的顺序，返回从右侧所能看到的节点值。

示例 1:
- 输入: [1,2,3,null,5,null,4]
- 输出: [1,3,4]

示例 2:
- 输入: [1,null,3]
- 输出: [1,3]

示例 3:
- 输入: []
- 输出: []

提示:
- 二叉树的节点个数的范围是 [0,100]
- -100 <= Node.val <= 100

#### 1. DFS
分析：找右视图，以右子树为主，可以先递归右子树，再递归左子树。注意，只有当 某一深度 的节点，右子树 不存在时，才将对应层数左节点加入结果列表。
如何判断节点是否记录到结果：
1. 在递归的同时记录 一个节点深度。
2. 如果递归深度 = 答案长度，则此节点加入答案。先右递归再左子树递归，确保右子树存在时，先加入结果。
3. 而当递归左子树时，只有当 出现 **新的深度** 时，左子树节点才会被 加入结果。

- 时间复杂度：O(n)
- 空间复杂度：O(h), h为树的高度


```python
from typing import Optional
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def rightSideView(self, root: Optional[TreeNode]) -> list[int]:
        ans = [] # 创建一个空列表，用于存储结果

        def dfs(node, depth): # 当前递归节点 和 已有深度
            if not node: # 如果当前节点为空，直接返回
                return

            if depth == len(ans): # 如果当前深度等于结果列表的长度，说明这是该深度的第一个节点
                ans.append(node.val) # 将当前节点的值添加到结果列表中

            dfs(node.right, depth + 1) # 先访问右子树
            dfs(node.left, depth + 1) # 再访问左子树
        dfs(root, 0)
        return ans # 返回结果列表
```

#### 2. BFS 层序遍历
层序遍历数中每一层，每层被视为一个队列，只将每层最后一个 节点添加到结果中。

步骤：套用102题目模板： **相关题目**：102. 二叉树的层序遍历
- 时间复杂度：O(n)； n 为二叉树的节点数。
- 空间复杂度：O(n)；

```python
# 层序遍历
class Solution:
    def rightSideView(self, root: Optional[TreeNode]) -> List[int]:
        if not root: 
            return []
        ans = []
        queue = [root]

        while queue:
            ans.append(queue[-1].val) # 将当前层的最后一个节点的值添加到结果列表中
            nxt = [] # 创建一个新的列表，用于存储下一层的节点
            for node in queue: # 遍历当前层的所有节点
                if node.left:
                    nxt.append(node.left) # 添加左子树
                if node.right:
                    nxt.append(node.right) # 添加右子树
            queue = nxt # 更新队列为下一层的节点
        return ans # 返回结果列表
```
