### 279. 完全平方数
给你一个整数 n ，返回 和为 n 的完全平方数的最少数量 。

完全平方数 是一个整数，其值等于另一个整数的平方；换句话说，其值等于一个整数自乘的积。例如，1、4、9 和 16 都是完全平方数，而 3 和 11 不是。

示例 1：
- 输入：n = 12
- 输出：3
- 解释：12 = 4 + 4 + 4

示例 2：
- 输入：n = 13
- 输出：2
- 解释：13 = 4 + 9

提示：`1 <= n <= 10^4`

#### 1.记忆化搜索（完全背包问题）
把 1,4,9,16,⋯ 这些**完全平方数视作物品体积**，物品**价值都是 1**。由于每个数（物品）选的**次数没有限制**，所以本题是一道标准的**完全背包问题**。求最小价值。
步骤：
1. 定义 dfs(i,j) 表示从前 i 个完全平方数中选一些数（可以重复选），满足元素和恰好等于 j，最少要选的数字个数。
2. 考虑第 i 个完全平方数 i2 选或不选：
   1. 不选：问题变成从前 i−1 个完全平方数中选一些数（可以重复选），满足元素和恰好等于 j。 `dfs(i, j) = dfs(i-1, j)`
   2. 选：问题变成从前 i 个完全平方数中选一些数（可以重复选），满足元素和恰好等于 j−i^2。 `dfs(i, j) = dfs(i, j-i^2) + 1`
3. 递归公式：取最小值 `dfs(i, j) = min(dfs(i-1, j), dfs(i, j-i^2) + 1)`
4. 递归边界：
   1. dfs(0,0)=0，因为没有数可以选了，且要得到的数等于 0，那么答案为 0。
   2. 如果 j>0，那么 `dfs(0,j)= ∞`，这里用 ∞ 表示不合法的状态，从而保证上式中的 min 取到合法的状态。
5. 本题一定有解，因为 1 是完全平方数。
6. 递归入口：由于 `i^2 ≤ n`，所以 i≤⌊ √n ​⌋，所以递归入口为 dfs(⌊ n​ ⌋, n)，也就是答案。

```python
from math import isqrt
from functools import cache
# 写在外面，多个测试数据之间可以共享，减少计算量
@cache
def dfs(i, j):
    if i == 0:
        return float('inf') if j else 0 # 无平方数
    if j < i*i: # 容量不够了,只能不选
        return dfs(i - 1, j)
    else:
        # 选 vs 不选 比较
        return min(dfs(i - 1, j), dfs(i, j - i*i) + 1)
class Solution:
    def numSquares(self, n: int) -> int:
        return dfs(isqrt(n), n)
    
n = 12
print(Solution().numSquares(n)) # 12 = 4 + 4 + 4
```

#### 2.将递归翻译为递推
可以去掉递归中的「递」，只保留「归」的部分，即自底向上计算。

具体来说，f[i][j] 的定义和 dfs(i,j) 的定义是一样的，都表示从前 i 个完全平方数中选一些数（可以重复选），满足元素和恰好等于 j，最少要选的数字个数。

相应的递推式为：
- 当 j < i^2 时：  `f[i][j] = f[i -1][j]`
- 当 j >= i^2 时： `f[i][j] = min( f[i -1][j], f[i][j - i * i] + 1)`
- 初始值 f[0][0]=0, f[0][j]=∞ (j>0).

```python
# 翻译成递推
N = 10000
dp = [[0] * (N+1) for _ in range(isqrt(N)+1)]
dp[0] = [0] + [float('inf')] * N

for i in range(1, len(dp)):
    for j in range(N+1): # 枚举背包容量
        if j < i*i: # 容量不够了,只能不选
            dp[i][j] = dp[i - 1][j]
        else:
            # 选 vs 不选 比较
            dp[i][j] = min(dp[i - 1][j], dp[i][j - i*i] + 1)

class Solution:
    def numSquares(self, n: int) -> int:
        return dp[isqrt(n)][n] # 或写为 dp[-1][n]，因为最后一行就是答案
    
n = 13
print(Solution().numSquares(n)) # 13 = 4 + 9
```
