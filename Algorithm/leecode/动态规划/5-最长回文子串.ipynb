### 5. 最长回文子串
给你一个字符串 s，找到 s 中最长的 回文 子串。

示例 1：

输入：s = "babad"
输出："bab"

解释："aba" 同样是符合题意的答案。

示例 2：

输入：s = "cbbd"
输出："bb"


#### 1. 中心扩散法
从某一中心开始向外左右扩散判断。
1. 奇数回文串
   1. 枚举 i=0,1,2,…,n−1 作为奇回文串的中心，向左右两侧扩展：
   2. 初始化 l=r=i。
   3. 如果 s[l]=s[r]，那么向左右两侧扩展，把 l 减一，把 r 加一，继续判断更长的子串是不是回文串。直到下标出界或者 `s[l] !=s[r]`。
   4. 循环结束时，最后一轮循环的子串 s[l+1] 到 s[r−1] 是回文串，若其长度 r−l−1 大于答案的长度，那么更新答案的左右端点为 l+1 和 r−1，方便输出具体子串。
2. 偶数回文串
   1. 枚举 i 和 i+1 作为偶回文串的中心，初始化 l=i，r=i+1。
   2. 其余做法同上。


```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        # 中心扩散法
        ans_left, ans_right = 0, 0 # 左右记录点

        n = len(s)
        # 奇数子串
        for i in range(len(s)):
            l = r = i
            while l >= 0 and r < n and s[l] == s[r]:
                l -= 1
                r += 1
            # 循环结束到 s[l] != s[r]
            if r - l -1 > ans_right - ans_left: # 新的最长回文串
                ans_left, ans_right = l + 1, r #左闭右开
                
        # 偶数子串
        for i in range(n-1):
            l, r = i, i + 1
            while l >= 0 and r < n and s[l] == s[r]:
                l -= 1
                r += 1
            if r - l - 1 > ans_right - ans_left:
                ans_left, ans_right = l + 1, r # 左闭右开
            
        return s[ans_left: ans_right]
```

#### 2. 动态规划
由于要判断边界 i, j 内的字符是否为回文串，动态规划中的状态采用记录采用二维数组dp[i][j]，dp[i][j] 表示 s[i:j] 是否为回文串。
1. 定义状态：dp[i][j] 表示 s[i:j] 是否为回文串。
2. 初始化边界条件：d[i][j] = false
3. 状态转移方程：
    - dp[i][j] =true;  s[i] == s[j] and dp[i + 1][j - 1] == true
    - dp[i][j] = true; i == j(仅一字符) and s[i] == s[j] 
    - dp[i][j] = true; j - i < 2(仅两个字符) and s[i] == s[j]
4. 更新回文串长度 max_len:
    - if dp[i][j] == true and j - i + 1 > max_len:  max_len = j - i + 1;  str = s[i: j]


```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        # 动态规划，由于回文串有两边界，状态记录采用二维数组
        n = len(s)
        if(n < 2):
            return s

        max_len = 0
        # 初始化状态
        dp = [[False] * n for _ in range(n)]
        left, right = 0, 0 # 记录回文串边界
        for i in range(n - 1, -1, -1):
            for j in range(i, n):
                # 判断为回文串
                if s[i] == s[j]:
                    # a, aa, aba, dp[i+1][j-1]
                    if j - i <= 2 or dp[i + 1][j - 1] == True:
                        dp[i][j] = True

                # 判断回文串长度更新
                if dp[i][j] == True and j - i + 1 > max_len:
                    max_len = j - i + 1
                    left, right = i, j + 1 # 左闭右开

        return s[left: right]
```
