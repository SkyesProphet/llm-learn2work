### 121.买股票的最佳时机
**动态规划入门**
给定一个数组 prices ，它的第 i 个元素 prices[i] 表示一支给定股票第 i 天的价格。

你只能选择 某一天 买入这只股票，并选择在 未来的某一个不同的日子 卖出该股票。设计一个算法来计算你所能获取的最大利润。

返回你可以从这笔交易中获取的最大利润。如果你不能获取任何利润，返回 0 。

#### 1.贪心算法
贪心算法：只买卖一次可用贪心，取最小左值和最大右值，求最大利润。

price为当天的价格，为了实现最大利润，设计变量 cost 记录前 i 天的最低价格，变量 profit 存储利润。

算法：
1. 由于初始值 i=0 ，为了序号对应，本文设从第 0 天开始；
2. 更新前 i 天的最低价格，即最低买入成本 cost；
3. 更新前 i 天的最高利润 profit ，即选择「前 i−1 天最高利润 profit 」和「第 i 天卖出的最高利润 price - cost 」中的最大值 ；


```python
from typing import List
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        cost, profile = float('inf'), 0 # 初始化成本和利润
        
        # 遍历价格列表，更新成本和利润
        for price in prices:
            cost = min(cost, price) # 更新当前天最低成本
            profile = max(profile, price - cost) # 更新当前天最大利润

        return profile

prices1 = [7,1,5,3,6,4]
print(Solution().maxProfit(prices1))# 贪心算法2： 逆序遍历
# 找出第 i 天之后的最高价格，即最优卖出
# 减去第 i 天价格，即第 i 天买入可获最大利润。
class Solution:
    def maxProfile(self, prices: List[int]) -> int:
        profile, max_price = 0, 0 # 初始化利润和最高价格
        
        # 逆序遍历价格列表，更新最高价格和利润
        for price in prices[::-1]: # 逆序遍历
            max_price = max(max_price, price) # 更新此天后最高价格
            profile = max(profile, max_price - price) # 更新此天买入可获最大利润
        return profile
    
prices1 = [7,1,5,3,6,4]
print(Solution().maxProfile(prices1))
prices2 = [7,6,4,3,1]
print(Solution().maxProfile(prices2))
prices2 = [7,6,4,3,1]
print(Solution().maxProfit(prices2))
```

#### 1. 动态规划
设计状态
1. 到最后交易结束时，一共会有3种状态，每种状态标识此状态下的最大利润：
   - dp0：一直不买
   - dp1：只买了一次
   - dp2：买了一次，卖了一次
2. 初始化三种状态，假设本金为 0
   - dp0 = 0
   - dp1 = -prices[0] # 买入，0 - prices[0]
   - dp2 = float('-inf') # 因为第一天不可能有卖出，故初始化置为负无穷
3. 状态转移方程
   1. dp0 = 0 # 不买，一直为 0
   2. dp1 = max(dp0 - prices[i], dp1) # 买入一次只有两种情况
      1. 前一天也是 dp1;
      2. 前一天是 dp0，今天买入：dp0 - prices[i]
   3. dp2 = max(dp1 + prices[i], dp2)
      1. 前一天是 dp1，今天卖出：dp1 + prices[i]
      2. 前一天是 dp2，今天不卖出，保持 dp2


```python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        # 初始化每天三种状态
        dp0 = 0  # 一直不买
        dp1 = -prices[0] # 只买了一次，初始化第首日
        dp2 = float('-inf') # 买了一次，卖了一次， 首日不可能卖出，利润设为负无穷
        for i in range(1, len(prices)):
            dp0 = 0 # 不买，现金一直为0
            dp1 = max(dp0 - prices[i], dp1) # dp1一般为负
            dp2 = max(dp1 + prices[i], dp2)

        return max(dp0, dp2)
```

动态规划2：
需要设置两个状态 dp[i][0]表示第i天持有股票时最大现金，dp[i][1]表示第i天不持有股票状况的最多现金。

算法步骤：
1. 以本金为0，买入 -price[i], 卖出 +price[i], 
2. 持有状态 dp[i][0] 现金为 负数， 卖出获得收益可为 正、负; 最后取dp[i][1]最大值,可为0（无买卖）、正（有买卖）.
3. 初始化：
4. 状态转移：
   1. dp[i][0] = max(dp[i-1][0], 0 - price[i])
      - 第 i 日持有 有两种可能性： 
      2. 保持持有：dp[i-1][0] 
      3. 第 i 日才买入：0 - price[i]
   2. dp[i][1] = max(dp[i-1][1], dp[i-1][0] + price[i])
      1. 第 i 日卖出：dp[i-1][0] + price[i] 
      2. 保持第 i- 1日不持有：dp[i-1][1]

```python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        l = len(prices)
        if l == 0:
            return 0
        # 初始化
        dp = [[0] * 2 for _ in range(l)] # 创建 l 行 2列的二维数组
        dp[0][0] = -prices[0] # 第一天买入，持有股票
        dp[0][1] = 0 # 第一天不持有股票，利润为0

        for i in range(1, l): # 遍历后续天数
            dp[i][0] = max(dp[i - 1][0], 0 - prices[i]) # 第 i 天持有股票的最大利润
            dp[i][1] = max(dp[i - 1][1], dp[i - 1][0] + prices[i]) # 第 i 天不持有股票的最大利润
        return dp[-1][1] # 最后一天不持有股票的最大利润
```
