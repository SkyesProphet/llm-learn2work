### 72. 编辑距离
给你两个单词 word1 和 word2， 请返回将 word1 转换成 word2 所使用的最少操作数  。

你可以对一个单词进行如下三种操作：
- 插入一个字符
- 删除一个字符
- 替换一个字符

示例 1：

输入：word1 = "horse", word2 = "ros"
输出：3

解释：
```
horse -> rorse (将 'h' 替换为 'r')
rorse -> rose (删除 'r')
rose -> ros (删除 'e')
```
示例 2：
输入：word1 = "intention", word2 = "execution"
输出：5
解释：
```
intention -> inention (删除 't')
inention -> enention (将 'i' 替换为 'e')
enention -> exention (将 'n' 替换为 'x')
exention -> exection (将 'n' 替换为 'c')
exection -> execution (插入 'u')
```
提示：
- 0 <= word1.length, word2.length <= 500
- word1 和 word2 由小写英文字母组成


#### 1.动态规划
递归搜索 + 保存计算结果 = 记忆化搜索

只需考虑从右往左（或者从左往右）操作:
假设两个字符为 s[i] 和 t[j]，先操作 i 再操作 j，还是先操作 j 再操作 i，结果都是一样的。推广，任意一种操作顺序，都可以重新排列成从右往左（从左往右）操作。


````python
class Solution:
    def minDistance(self, word1: str, word2: str) -> int:
        n, m = len(word1), len(word2)

        @cache
        def dfs(i:int, j:int) -> int:
            # 从右往左遍历
            if i < 0: #遍历完字符串则回退到 0
                return j + 1
            if j < 0:
                return i + 1
            if word1[i] == word2[j]: #字符相等，同时往前走 1
                return dfs(i - 1, j - 1)
            return min(dfs(i - 1, j), dfs(i, j - 1), dfs(i - 1, j - 1)) + 1 # +1是操作

        return dfs(n - 1, m - 1)
````

````python
class Solution:
    def minDistance(self, word1: str, word2: str) -> int:
        # 动态规划初始化
        dp = [[0]* (len(word2) +1) for _ in range(len(word1) +1)]

        for i in range(len(word1) +1): # dp[0][0] = 0
            dp[i][0] = i
        for j in range(len(word2) +1):
            dp[0][j] = j

        # 状态变换，只需考虑删除和替换，一字符串增加操作等于另一字符串删除操作
        for i in range(1, len(word1) +1):
            for j in range(1, len(word2) +1):
                if word1[i-1] == word2[j-1]:
                    dp[i][j] = dp[i-1][j-1]
                else:
                    dp[i][j] = min(dp[i-1][j-1], dp[i-1][j], dp[i][j-1]) + 1 # 分别对应替换，删word1[i-1],删word2[j-1]

        return dp[-1][-1]

        
````
