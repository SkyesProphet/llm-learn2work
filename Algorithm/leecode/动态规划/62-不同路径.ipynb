### 62. 不同路径
一个机器人位于一个 m x n 网格的左上角 （起始点在下图中标记为 “Start” ）。

机器人每次只能向下或者向右移动一步。机器人试图达到网格的右下角（在下图中标记为 “Finish” ）。

问总共有多少条不同的路径？
示例 1：
输入：m = 3, n = 7
输出：28

示例 2：
输入：m = 3, n = 2
输出：3

解释：
从左上角开始，总共有 3 条路径可以到达右下角。
1. 向右 -> 向右 -> 向下
2. 向右 -> 向下 -> 向右
3. 向下 -> 向右 -> 向右

示例 3：
输入：m = 7, n = 3

#### 1. 动态规划
##### 1.1 问题分析
假设左上角的坐标是 (0,0)，右下角的坐标是 (2,6)。

最后一步发生了什么？
- 如果从 (1,6) 向下走到终点 (2,6)，那么要解决的问题是从起点 (0,0) 走到 (1,6) 的路径数。
- 如果从 (2,5) 向右走到终点 (2,6)，那么要解决的问题是从起点 (0,0) 走到 (2,5) 的路径数。
这些问题都是和原问题相似的、规模更小的子问题，可以用递归解决。

##### 1.2 状态转移方程
定义状态为 dfs(i,j)，表示从起点 (0,0) 走到 (i,j) 的路径数。

讨论如何到达(i,j) 的：
- 如果是从 (i−1,j) 过来，那么问题变成从起点 (0,0) 走到 (i−1,j) 的路径数，即 dfs(i−1,j)。
- 如果是从 (i,j−1) 过来，那么问题变成从起点 (0,0) 走到 (i,j−1) 的路径数，即 dfs(i,j−1)。
- 状态转移方程为： dfs(i,j) = dfs(i−1,j) + dfs(i,j−1)

递归边界：
- dfs(−1,j)=dfs(i,−1)=0。无法从 (0,0) 到达这些位置。
- dfs(0,0)=1。起点到它自己有一条路径，即原地不动。
- 递归入口：dfs(m−1,n−1)，即finish的位置

##### 1.3 递归搜索 + 保存递归返回值 = 记忆化搜索
考虑到整个递归过程中有大量重复递归调用（**递归入参相同**）。由于递归函数没有副作用，同样的入参无论计算多少次，算出来的结果都是一样的，因此可以用记忆化搜索来优化：
- 如果一个状态（递归入参）是第一次遇到，那么可以在返回前，把状态及其结果记到一个 memo 数组中。
- 如果一个状态不是第一次遇到（memo 中保存的结果不等于 memo 的初始值），那么可以直接返回 memo 中保存的结果。

注意：memo 数组的初始值一定不能等于要记忆化的值！例如初始值设置为 0，并且要记忆化的 dfs(i,j) 也等于 0，那就没法判断 0 到底表示第一次遇到这个状态，还是表示之前遇到过了，从而导致记忆化失效。**一般把初始值设置为 −1**。本题由于方案数不为 0，也可以初始化成 0。



```python
# 动态规划：递归法
class Soluition:
    def uniquePaths(self, m: int, n: int) -> int:
        @cache # 缓存装饰器，避免重复计算 dfs 的结果（一行代码实现记忆化）
        def dfs(i, j) -> int:
            if i < 0 or j < 0: #越界
                return 0
            
            if i == 0 and j == 0: # 最初节点
                return 1
            return dfs(i - 1, j) + dfs(i, j - 1) # 上方节点 + 左方节点
        return dfs(m - 1, n - 1) # 从终点开始回溯
```

```python
# 动态规划：递推法
class Soluition:
    def uniquePaths(self, m: int, n: int) -> int:
        mem = [[0] * (n + 1) for _ in range(m + 1)]  # 初始化二维数组
        for i in range(m):
            for j in range(n): #这里默认 i,j比实际位置小
                if i ==  j == 0:  # 边界条件
                    mem[1][1] = 1
                else:
                    mem[i + 1][j + 1] = mem[i][j +1] + mem[i + 1][j]
            return mem[m][n]
```
