### 438. 找到字符串中所有字母异位词
给定两个字符串 s 和 p，找到 s 中所有 p 的 **异位词** 的子串，返回这些子串的起始索引。不考虑答案输出的顺序。

示例 1:

输入: s = "cbaebabacd", p = "abc"
输出: [0,6]

解释:
- 起始索引等于 0 的子串是 "cba", 它是 "abc" 的异位词。
- 起始索引等于 6 的子串是 "bac", 它是 "abc" 的异位词。

示例 2:

输入: s = "abab", p = "ab"
输出: [0,1,2]

解释:
- 起始索引等于 0 的子串是 "ab", 它是 "ab" 的异位词。
- 起始索引等于 1 的子串是 "ba", 它是 "ab" 的异位词。
- 起始索引等于 2 的子串是 "ab", 它是 "ab" 的异位词。

#### 1. 滑动窗口(定长)—— 找异位词子串（长度固定）
枚举 s 的所有长为 n 的子串 t，如果 t 的每种字母的出现次数，和 p 的每种字母的出现次数都相同，那么 t 是 p 的异位词。

1. 用滑动窗口枚举 s 的所有长为 n 的子串 t。
2. 维护 t 的每种字母的出现次数。
3. 对比 t, p的每种字母出现的次数，如果相同，则 t 是 p 的异位词。
4. 把 t 左端点下标加入答案。

时间复杂度：O(m+n): m为s的长度，n为p的长度

```python
from typing import List
from collections import Counter
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        cnt_p = Counter(p) # 统计 p 中每种字母的频次
        cns_s = Counter() # 初始化hash表，统计 s 的长为 len(p) 的子串 t 的每种字母频次

        ans = []

        # 定长滑动窗口
        for right, c in enumerate(s):
            cns_s[c] += 1 # 将 s 中当前字符加入 hash 表

            left = right - len(p) + 1 # 维护定长窗口左边界
            if left < 0: # 当窗口长度不足 len(p) 时，继续向右扩展
                continue

            if cns_s == cnt_p: # 判断 hash 表是否相等
                ans.append(left) # 左边界索引加入结果

            cns_s[s[left]] -= 1 # 下一次移动前，将左边界字符从 hash 表中移除

        return ans

s = "cbaebabacd"
p = "abc"
print(Solution().findAnagrams(s, p))
```

#### 2. 不定长滑窗
枚举子串 t 的右端点。
- 如果发现 t 其中一种字母的出现次数大于 p 的这种字母的出现次数，则增大 t 的左端点（缩小窗口）。
- 如果发现 t 的长度等于 p 的长度，则说明 t 的每种字母的出现次数，等于 p 的每种字母的出现次数，即 t 是 p 的异位词。
- 内层循环中，t 的每种字母的出现次数，都小于等于 p 的每种字母的出现次数。
  - 如果 t 的其中一种字母的出现次数比 p 的小，那么 t 的长度必然小于 p 的长度。
  - 只要 t 的长度等于 p 的长度，则 t 就是 p 的异位词。

适用 cnt 数组记录 p, t 中每种字母出现的次数之 **差** 。
- 统计 p 的字母 c， cnt[c]++。先统计完p，再统计 t，进行加减操作。
- 对于 t 的字母 c，右端点滑动，t 中字符加入 cnt[c]--。左端点滑动，t字符减少 cnt[c]++。
- 如果 cnt[c]<0，说明窗口中的字母 c 的个数比 p 的多，右移左端点。

时间复杂度：O(m)

```python
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        cnt = Counter(p) # 统计 p 中每种字母的频次
        ans = []

        left = 0
        # cnt 记录 p, t 字符频次差异
        for right, c in enumerate(s):
            cnt[c] -= 1     # 右端点字母进入窗口，差异减少

            while cnt[c] < 0:  # 字母 c 太多了
                cnt[s[left]] += 1  # 左端点字母离开窗口
                left += 1   #缩小窗口
            
            # 由于维护窗口 t 中字符 <= p，故相等时
            if right - left + 1 == len(p): # t 和 p 中字符频次完全相同
                ans.append(left)
        return ans
```
