### 560. 和为 K 的子数组
给你一个整数数组 nums 和一个整数 k ，请你统计并返回 该数组中和为 k 的子数组的个数 。

子数组是数组中元素的连续非空序列。

示例 1：

输入：nums = [1,1,1], k = 2
输出：2

示例 2：

输入：nums = [1,2,3], k = 3
输出：2

提示：

- `1 <= nums.length <= 2 * 104`
- `-1000 <= nums[i] <= 1000`
- `-107 <= k <= 107`

#### 相关知识：前缀和
**前缀和**思路指：
1. 一般地，任意子数组都是一个前缀去掉前缀后的结果。所以**任意子数组的和，都可以表示为两个前缀和的差**。
2. 比如 `nums=[1,2,3,4,5,6]`，要想计算子数组 `[3,4,5]` 的元素和，可以用前缀 `[1,2,3,4,5]` 的元素和，减去另一个前缀 `[1,2]` 的元素和，就得到了子数组 `[3,4,5]` 的元素和。
3. 预处理 nums 所有的前缀和，就可以 O(1) 计算任意子数组的元素和。
4. 前缀和计算：
   1. `s[0]=0`
   2. `s[i + 1] = s[i] + nums[i]` # 即，计算当前位置数 的 前缀的和
   3. 注意 s 是一个长为 n+1 的数组，第一个数是 0。
   
**问题转换**：
1. 设 `i< j`，如果 `nums[i]` 到 `nums[j−1]` 的元素和等于 k，用前缀和表示，就是:`s[j] − s[i] = k`。
2. 求 s 中有多少对下标 (i,j) 满足 `0 ≤ i < j ≤ n` 且 `s[j] − s[i] = k`?


```python
# 前缀和计算的实现
class NumArray:
    def __init__(self, nums):
        self.prefix_sum = [0] * (len(nums) + 1) # 前缀和数组，长度比输入数组多1
        for i, x in enumerate(nums):
            self.prefix_sum[i + 1] = self.prefix_sum[i] + x
    
    def sumRange(self, left, right):
        return self.prefix_sum[right + 1] - self.prefix_sum[left]
            
```

#### 1. 枚举右，维护左
用二重循环暴力枚举有多少个 `s[j]−s[i] = k`，时间复杂度是 O(n^2)，太慢了。

问题转换：
1. `s[j] - s[i] = k` 可移项为 `s[i] = s[j] -k`
2. 枚举当前的前缀和 `s[j]`，看看曾经有多少个前缀和等于 `s[j]−k`（配对）。
3. 即在位置 j 的前缀序列中 寻找 值为 `s[j] - k`的个数。 
实现：
1. 遍历 s[j] 的同时，用一个哈希表 cnt 统计，key 是 s[j]，value 是值为 s[j] 的前缀和的个数。
2. 统计哈希表中的数。

问：为什么要把 s[0]=0 也加到哈希表中？

答：举个最简单的例子，nums=[1], k=1。如果不把 s[0]=0 加到哈希表中，按照我们的算法，没法算出这里有 1 个符合要求的子数组。也可以这样理解，要想把任意子数组都表示成两个前缀和的差，必须添加 s[0]=0，否则当子数组是前缀时，没法减去一个数

链接：https://leetcode.cn/problems/subarray-sum-equals-k/solutions/2781031/qian-zhui-he-ha-xi-biao-cong-liang-ci-bi-4mwr/

```python
from typing import List
from collections import defaultdict
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        s = [0] * (len(nums) + 1)

        # 计算前缀和
        for i, x in enumerate(nums):
            s[i + 1] = s[i] + x
        
        ans = 0
        cnt = defaultdict(int) # 可处理键不存在的情况
        for sj in s:
            ans += cnt[sj - k] # sj - (sj - k) = k
            cnt[sj] += 1
        return ans
    
nums = [1,1,1]
k = 2
print(Solution().subarraySum(nums, k))
```

```python
#写法优化——一次遍历
# 可以一边计算前缀和，一边遍历前缀和。
# 遍历 nums 前，先统计 s[0]=0，即空前缀的元素和等于 0。往 cnt 中添加 cnt[0]=1。

class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        ans = 0
        cnt = defaultdict(int)
        cnt[0] = 1 # 统计空前缀的元素和等于 0
        s = 0 # 前缀和
        for x in nums:
            s += x
            ans += cnt[s - k] # sj - (sj - k) = k
            cnt[s] += 1
        return ans
    
nums = [1,2,3]
k = 3
print(Solution().subarraySum(nums, k))
```
