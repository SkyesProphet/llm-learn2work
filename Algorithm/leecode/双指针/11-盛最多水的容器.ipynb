### 11.盛最多水的容器
给定一个长度为 n 的整数数组 height 。有 n 条垂线，第 i 条线的两个端点是 (i, 0) 和 (i, height[i]) 。

找出其中的两条线，使得它们与 x 轴共同构成的容器可以容纳最多的水。

返回容器可以储存的最大水量。

示例 1：

输入：height = [1,8,6,2,5,4,8,3,7]
输出：49

解释：图中垂直线代表输入数组 [1,8,6,2,5,4,8,3,7]。在此情况下，容器能够容纳水（表示为蓝色部分）的最大值为 49。

示例 2：

输入：height = [1,1]
输出：1

#### 1. 双指针
由于 容器容积受到 横向宽 和最短的柱子高 影响。
1. 横向宽：左右边界，即 i 和 j
2. 最短的柱子高： `height[i] if height[i] < height[j] else height[j]`

步骤
1. 取 左右指针作为边界，i < j， 初始化界为 i = 0, j = n-1
2. 移动 i 或 j，取最短的柱子高，计算当前容积，并更新记录最大容积
3. 每次移动，判断当前柱子高，如果 i < j，则 i++，否则 j--

```python
from typing import List
class Solution:
    def maxArea(self, height: List[int]) -> int:
        # 双指针 i, j首尾相中间移动，计算面积，保存更新最大面积
        i, j = 0, len(height) - 1
        ans = 0
        # 若左右指针未相遇，可构成容器
        while i < j:
            ans = max(ans, (j - i) * min(height[i], height[j]))
            if height[i] < height[j]:
                i += 1
            else:
                j -= 1
        return ans
    
height = [1,8,6,2,5,4,8,3,7]
print(Solution().maxArea(height))
```
