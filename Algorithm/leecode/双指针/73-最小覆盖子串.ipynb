### 76.最小覆盖子串
**知识点：滑动窗口**
给定两个字符串 s 和 t，长度分别是 m 和 n，返回 s 中的 **最短窗口** 子串，使得该子串包含 t 中的每一个字符（包括重复字符）。如果没有这样的子串，返回空字符串 ""。
**s 子串中字符出现次数必须不少于 t 中对应字符的次数**。

示例 1：
输入：s = "ADOBECODEBANC", t = "ABC"
输出："BANC"

示例 2：
输入：s = "a", t = "a"
输出："a"

示例 3：
输入：s = "a", t = "aa"
输出：""
解释：t 中两个字符 'a' 均应包含在 s 的子串中，
因此没有符合条件的子字符串，返回空字符串。


#### 1.滑动窗口
1. 利用Counter函数统计字符串t中所需字符数
   1. Counter是 collections 模块中的一个字典子类。
   2. 统计可迭代对象中每个元素的出现次数，并返回一个类似字典的结构。
2. 初始化窗口左右指针 l, r; 遍历 s， 移动 r 直到窗口中包含所有 t 中的字符
3. 找齐之后开始收缩窗口，右移 l, 直到子串窗口有字母出现次数小于t 中对应字母的次数
4. 保存覆盖子串，多次判断出最小的覆盖子串

```python
from collections import Counter
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        # Counter 统计字符串中每个字符出现的次数,返回dict
        cnt_s = Counter() # 初始化子串统计
        cnt_t = Counter(t) # 目标字符串统计

        ans_left, ans_right = -1, len(s) # 记录最优解左右边界
        left = 0 # 滑动窗口左边界

        for right, c in enumerate(s): # 右边界向右 0 到 n-1移动
            cnt_s[c] += 1 # 右边界 s[right] 加入子串统计

            # 判断当前子串是否包含 t 中所有字符
            while cnt_s >= cnt_t: 
                # 1. 更新最优解:找到更短的子串
                if right - left < ans_right - ans_left:
                    ans_left, ans_right = left, right
                # 2. 左边界向右移动，缩小子串
                cnt_s[s[left]] -= 1 # 左边界 s[left] 从子串统计中移除
                left += 1 # 左边界右移
                # 直到当前子串不再包含 t 中所有字符，继续右边界移动
        if ans_left < 0:
            return "" # 没有找到满足条件的子串
        else:
            return s[ans_left:ans_right + 1] # 左闭右开，不包含 ans_right + 1
```

```python
# 优化速度
from collections import defaultdict
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        # 注：defaultdict 比 Counter 快
        diff = defaultdict(int)  # 窗口每种字母个数 - t 每种字母个数
        for c in t:
            diff[c] -= 1
        kinds = len(diff)  # t 中有 kinds 种不同的字母

        ans_left, ans_right = -1, len(s)
        ge_cnt = 0  # 窗口内有 ge_cnt 种字母的出现次数 >= t 中相应字母的出现次数
        left = 0

        for right, c in enumerate(s):  # 移动子串右端点
            diff[c] += 1  # 右端点字母移入子串
            if diff[c] == 0:  # 原来窗口内 c 的出现次数比 t 的少，现在一样多
                ge_cnt += 1  # 从 < 变成 >=

            while ge_cnt == kinds:  # 涵盖：所有字母的出现次数都是 >=
                if right - left < ans_right - ans_left:  # 找到更短的子串
                    ans_left, ans_right = left, right  # 记录此时的左右端点

                x = s[left]  # 左端点字母
                if diff[x] == 0:
                    # x 移出窗口之前，检查出现次数，
                    # 如果窗口内 x 的出现次数和 t 一样，
                    # 那么 x 移出窗口后，窗口内 x 的出现次数比 t 的少
                    ge_cnt -= 1  # 从 >= 变成 <
                diff[x] -= 1  # 左端点字母移出子串
                left += 1

        return "" if ans_left < 0 else s[ans_left: ans_right + 1]

# 作者：灵茶山艾府
# 链接：https://leetcode.cn/problems/minimum-window-substring/solutions/2713911/liang-chong-fang-fa-cong-o52mn-dao-omnfu-3ezz/
```
