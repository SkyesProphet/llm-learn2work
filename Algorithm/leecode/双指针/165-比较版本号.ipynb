### 165.比较版本号
给你两个 版本号字符串 version1 和 version2 ，请你比较它们。版本号由被点 '.' 分开的修订号组成。修订号的值 是它 转换为整数 并忽略前导零。

比较版本号时，请按 从左到右的顺序 依次比较它们的修订号。如果其中一个版本字符串的修订号较少，则将缺失的修订号视为 0。

返回规则如下：
- 如果 version1 < version2 返回 -1，
- 如果 version1 > version2 返回 1，
- 除此之外返回 0。

例 1：

输入：version1 = "1.2", version2 = "1.10"
输出：-1

解释：
version1 的第二个修订号为 "2"，version2 的第二个修订号为 "10"：2 < 10，所以 version1 < version2。

示例 2：

输入：version1 = "1.01", version2 = "1.001"
输出：0

解释：
忽略前导零，"01" 和 "001" 都表示相同的数字 "1"。

示例 3：

输入：version1 = "0.1", version2 = "1.1"
输出：-1

解释：version1 中第一个修订号是 "0"，version2 中第一个修订号是 "1"：0 < 1，所以 version1

#### 解法：切割，转换列表
1. 把字符串按照` . `分割，分割出的子串转成整数，我们可以得到两个整数列表。
2. 短的列表在末尾补 0。
3. 比较这两个列表的字典序：从左到右遍历两个列表（分别记作 a 和 b）。
   1. if a[i] < b[i]: return -1
   2. if a[i] > b[i]: return 1
   3. 否则，继续比较下一个元素。
   4. 如果 a 和 b 的元素都比较完，返回 0。 


```python
from itertools import zip_longest
class Solution:
    def compareVersion(self, version1: str, version2: str) -> int:
        a = list(map(int, version1.split('.')))
        b = list(map(int, version2.split('.')))

        for v1, v2 in zip_longest(a, b, fillvalue=0):
            if v1 != v2:
                return -1 if v1 < v2 else 1
        return 0
    
#version1 = "1.01"
#version2 = "1.001"
#version1 = "1.2"
#version2 = "1.10"
version1 = "1.0"
version2 = "1.0.0.0"
print(Solution().compareVersion(version1, version2))
```
