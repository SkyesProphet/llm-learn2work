#### 239. 滑动窗口最大值
给你一个整数数组 nums，有一个大小为 k 的滑动窗口从数组的最左侧移动到数组的最右侧。你只可以看到在滑动窗口内的 k 个数字。滑动窗口每次只向右移动一位。

返回 滑动窗口中的最大值 。


示例 1：

输入：nums = [1,3,-1,-3,5,3,6,7], k = 3
输出：[3,3,5,5,6,7]

示例 2：

输入：nums = [1], k = 1
输出：[1]

提示：
- `1 <= nums.length <= 105`
- `-104 <= nums[i] <= 104`
- `1 <= k <= nums.length`

#### 1.单调队列
1. 右边入（元素进入队尾，同时维护队列单调性）
   - 若新元素大，删除所有小于的老元素，
   - 由于队尾 新元素大时，会移除所有旧的小于元素，故单调队列一直降序。
2. 左边出（元素离开队首）
3. 记录/维护答案（根据队首）

注意：
为什么代码中的「队首离开窗口」写的是 if 而不是 while？
- 因为窗口每次向右滑动 1 位，所以元素也是 一个一个离开窗口的。

**步骤**
1. 初始化： 双端队列 deque ，结果列表 res ，数组长度 n；deque 内 仅包含窗口内的元素。
2. 滑动窗口： 左边界范围 `i∈[1−k,n−k]` ，右边界范围 `j∈[0,n−1]`;
   1. 若 `i>0` 且 队首元素 `deque[0] = 被删除元素 nums[i−1]` ：则队首元素出队；
   2. 删除 deque 内所有 < nums[j] 的元素，以保持 deque 递减；
   3. 将 nums[j] 添加至 deque 尾部；
   4. 若已形成窗口（即 i≥0 ）：将窗口最大值（即队首元素 deque[0] ）添加至列表 res ；

```python
from typing import List
from collections import deque
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        deq = deque()
        res, n = [], len(nums)

        # 
        for i, j in zip(range(0 - k +1, n - k + 1), range(n)):
            # Remove the elements which are out of this window
            if i > 0 and deq[0] == nums[i - 1]: # 满足左边界条件，且队首元素等于左边界元素
                deq.popleft() # 队首出队
            
            # 保持 deq 递减
            while deq and deq[-1] < nums[j]: # 队尾元素小于当前元素
                deq.pop() # 队尾出队

            deq.append(nums[j]) # 右指针当前元素入队
            # 记录当前窗口的最大值
            if i >= 0: # 满足左边界条件
                res.append(deq[0]) # 队首元素为当前窗口的最大值
        return res
                        
nums = [1,3,-1,-3,5,3,6,7]
k = 3
print(Solution().maxSlidingWindow(nums, k))
```

```python
# 队列保存坐标的另一种写法
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        ans = [0] * (len(nums) - k + 1) # 窗口个数
        q =deque() # 存储元素索引

        for i, x in enumerate(nums): # 枚举窗口的右边界
            # 1. 右边入
            while q and nums[q[-1]] < x: # 弹出左侧比当前元素小的元素
                q.pop()
            q.append(i) # 添加当前元素

            # 2.左边出
            left = i - k + 1 # 窗口的左边界
            if q[0] < left: # 队首元素索引小于左边界索引
                q.popleft() # 队首出队
            
            # 3. 记录结果
            if left >= 0: # 窗口的左边界在数组范围内
                ans[left] = nums[q[0]] # 队首元素为当前窗口的最大值
        return ans
            
nums = [1,3,-1,-3,5,3,6,7]
k = 3
print(Solution().maxSlidingWindow(nums, k))
```
