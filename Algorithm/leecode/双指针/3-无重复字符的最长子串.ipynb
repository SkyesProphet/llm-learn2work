### 3. 无重复字符的最长子串
**知识点：滑动窗口**
滑动窗口相当于在维护一个队列。右指针的移动可以视作**入队**，左指针的移动可以视作**出队**。

题目：给定一个字符串 s ，请你找出其中不含有重复字符的 最长 子串 的长度。

示例 1:
输入: s = "abcabcbb"
输出: 3
解释: 因为无重复字符的最长子串是 "abc"，所以其长度为 3。

示例 2:
输入: s = "bbbbb"
输出: 1
解释: 因为无重复字符的最长子串是 "b"，所以其长度为 1。

示例 3:
输入: s = "pwwkew"
输出: 3
解释: 因为无重复字符的最长子串是 "wke"，所以其长度为 3。
请注意，你的答案必须是 子串 的长度，"pwke" 是一个子序列，不是子串。

#### 1.滑动窗口 + 哈希表
滑动窗口：双指针
哈希表：字典 dict 记录字符 `s[r]` **最后一次出现的索引**。
- 更新左指针 l ： 根据上轮左指针 l 和 dic[s[r]] ，每轮更新左边界 l ，保证区间 [l+1,r] 内无重复字符且最大。
  
  $l = max(l, dic[s[r]])$
- 更新结果 res ： 取上轮 res 和本轮双指针区间 [l+1,r] 的宽度（即 r−l ）中的最大值。


```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        # 滑动窗口 也是双指针法，依次固定左起始，判断右边界内是否有重复字符
        # 单纯的双指针法相当于暴力枚举
        # 优化的滑动窗口:
        # 在遇到重复字符时，此时区间存在字符出现两次，移动左边界，使窗口内不含重复字符。
        if len(s) < 2: # 0, 1
            return len(s)

        dic_index = {} # dict() 存储左指针下标
        l, res = -1, 0
        for r in range(len(s)):
            if s[r] in dic_index: # 存在重复字符
                l = max(dic_index[s[r]], l)  # 更新左指针位置，max()保证左指针只向右移动

            dic_index[s[r]] = r # 更新字符下标
            res = max(res, r - l)

        return res
```
