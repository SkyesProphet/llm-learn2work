### 283. 移动零
给定一个数组 nums，编写一个函数将所有 0 移动到数组的末尾，同时保持非零元素的相对顺序。

请注意 ，必须在不复制数组的情况下原地对数组进行操作。

示例 1:

输入: nums = [0,1,0,3,12]
输出: [1,3,12,0,0]

示例 2:

输入: nums = [0]
输出: [0]

#### 1.把 nums 当作栈

1. 用一个栈记录非零元素。
2. 非0数不入栈，记录 0 的数量，最后在栈后添加相应数量的 0.
   1. 为了做到 O(1) 空间复杂度，直接把 nums 当作栈，用一个变量 stackSize 表示栈的大小，初始值为 0。
   2. 入栈就是把 `nums[stackSize]` 置为 `nums[i]`，然后把 stackSize 加一。
   3. 最后把 nums 中的下标从 stackSize 到 n−1 的数都置为 0。

```python
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        s_size = 0 # stack_size
        for x in nums:
            if x != 0: # 遇到非0数，入栈
                nums[s_size] = x 
                s_size += 1
        for i in range(s_size, len(nums)): # 栈内元素全部入栈后，剩余位置全部置0
            nums[i] = 0
```

#### 2.双指针+交换元素
方法一在最坏情况下（nums 全为 0），需要遍历 nums 两次。能否做到一次遍历？

把 0 视作空位。将所有非零元素都移到数组左边的空位上，并保证非零元素的顺序不变。

即交换 0 与 后续非零数位置，为了保证非零元素的顺序不变，我们需要维护最左边的空位的位置（下标）。

**步骤**：
1. 从左到右遍历 `nums[i]`。
2. 维护另一个下标 i0（初始值为 0），并保证下标区间 `[i0​, i−1]` 都是空位，且 i0指向最左边的空位。
3. 如果 `nums[i]` 不为 0，则交换 `nums[i]` 和 `nums[i0]`，并更新 `i += 1, i0 += 1`。
4. 如果 `nums[i]=0`，无需交换，只把 i 加一。
