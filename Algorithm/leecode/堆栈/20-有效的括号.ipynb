### 20. 有效的括号
给定一个只包括 '('，')'，'{'，'}'，'['，']' 的字符串 s ，判断字符串是否有效。

有效字符串需满足：

1. 左括号必须用相同类型的右括号闭合。
2. 左括号必须以正确的顺序闭合。
3. 每个右括号都有一个对应的相同类型的左括号。

示例 1：

输入：s = "()"
输出：true

示例 2：

输入：s = "()[]{}"
输出：true

示例 3：

输入：s = "(]"
输出：false

示例 4：

输入：s = "([)]"
输出：false

示例 5：

输入：s = "{[]}"


#### 1. 括号匹配 栈 辅助
利用栈的先入后出的特性，将左括号入栈，右括号出栈，匹配成功则出栈，匹配失败则返回False。

构建哈希表 dict, 存储左括号和右括号之间的映射关系。方便查询。

算法流程：
1. 创建一个栈 stack，用于存储左括号。
2. 遍历字符串 s 中的每一个字符 c。
    1. 如果 c 是左括号，将其入栈。
    2. 如果 c 是右括号，判断栈是否为空，如果不为空，则将栈顶的左括号弹出，并判断弹出的左括号是否与 c 匹配。如果不匹配，则返回 False。
    3. 如果 c 是右括号，栈为空，则返回 False。

```python
class Solution:
    def isValid(self, s: str) -> bool:
        # 构建栈，左括号入，右括号出，匹配则出栈
        # 全匹配完 栈 为空，否则False

        # 简单计算长度
        if len(s) % 2 != 0:
            return False
        dic = {'(': ')', '[': ']', '{': '}'}
        stack = []
        for c in s: # 遍历字符串
            if c in dic: # 是左括号
                stack.append(c) # 入栈
            elif not stack or dic[stack[-1]] != c: #右括号 栈为空或者不匹配
                return False
            else: # 匹配成功
                stack.pop() # 出栈
        # 遍历完成
        return len(stack) == 0
```

```python
# 写法二
class Solution:
    def isValid(self, s: str) -> bool:
        # 使用栈，遇到左边，右边入栈，遇到右边出栈
        # 全匹配完栈为空，则True,否则False

        # 1、 数量不匹配
        if len(s) % 2 != 0:
            return False # 如果 s 的长度为奇数，一定不符合要求
        # 2、出栈入栈
        stack = []
        for char in s:
            if char == '(':
                stack.append(')')
            elif char == '[':
                stack.append(']')
            elif char == '{':
                stack.append('}')
            elif not stack or stack[-1] != char: # 遇到右括号，若栈为空或不等于栈顶
                return False
            else:
                stack.pop() # 若满足以上所有，即右括号配对了

        return True if not stack else False # 最终栈为空则True,否则False
```
