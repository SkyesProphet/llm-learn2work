### 216.组合总和Ⅲ
找出所有相加之和为 n 的 k 个数的组合，且满足下列条件：

只使用数字1到9
每个数字 最多使用一次 
返回 所有可能的有效组合的列表 。该列表不能包含相同的组合两次，组合可以以任何顺序返回。

**关联题**：77.组合 从后往前遍历，新增约束条件 n

示例1：

输入: k = 3, n = 7
输出: [[1,2,4]]

解释: 1 + 2 + 4 = 7

#### 1. 回溯——枚举选哪一个
由于限定数字范围为1-9，且要求数字不重复，因此可以使用回溯法
从后往前遍历， i 即为余下数个数，和当前数的 值
限制范围， t 为还需要的 数值和
1. 剪枝： d = k - len(path) # 剩余可选数
   1. 当 `t < 0` 时，此时组合数大于目标。剪枝
   2. 当 ` t > (i + i -d +1) *d // 2` 时，此时组合数小于目标, 已无法满足目标。剪枝
2. 递归终止条件： len(path) == k, 
   - 此时 d == 0, 跳过剪枝条件，说明 t > 0 和 t < 0都不满足，即 t == 0。即达成条件。 

```python
from typing import List
class Solution:
    def combinationSum3(self, k: int, n: int) -> List[List[int]]:
            # 回溯
            # 枚举9-1选哪一个，target为目标限制
            ans = []
            path = []

            def dfs(i , t):
                d = k - len(path)
                # 剪枝：如果剩余的数的个数小于d，就没有必要继续了
                if i < d:
                    return
                # 剪枝：如果当前组合大于n, 即t < 0,  或剩余的数的和小于t，就没有必要继续了
                if t < 0 or t > (i + i - d + 1) * d // 2:
                    return
                
                if len(path) == k: # # d=0,对比剪枝判断，说明t=0 满足
                     ans.append(path.copy())
                     return
                
                # 遍历枚举选哪个
                for j in range(i, 0, -1):
                    path.append(j)
                    dfs(j-1, t-j)
                    path.pop()
            dfs(9, n)
            return ans

k = 3
n = 7
print(Solution().combinationSum3(k, n))
```

#### 2.回溯——选或不选
依次判断条件，实现 选 or 不选
1. 不选： 需满足 i > d 条件; 当余下数不足时，必须选，只能跳过不选
2. 选， 继续下一个位置遍历 选择

```python
class Solution:
    def combinationSum3(self, k: int, n: int) -> List[List[int]]:
        # 选与不选
        ans = []
        path = []

        def dfs(i, t):
            d = k - len(path)
            if t < 0 or t > (i + i - d + 1) * d // 2:
                return
            if d == 0:
                ans.append(path.copy())
                return
            # 不选i
            if i > d:
                dfs(i-1, t)
            # 选i
            path.append(i)
            dfs(i-1, t-i)
            path.pop()
        
        dfs(9, n)
        return ans
k = 3
n = 7
print(Solution().combinationSum3(k, n))
```
