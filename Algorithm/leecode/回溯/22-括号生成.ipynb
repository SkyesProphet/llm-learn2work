### 22.括号生成

数字 n 代表生成括号的对数，请你设计一个函数，用于能够生成所有可能的并且 有效的 括号组合。


示例 1：

输入：n = 3
输出：["((()))","(()())","(())()","()(())","()()()"]

示例 2：

输入：n = 1
输出：["()"]

**思路**：本题目本质 **77.组合** 问题，只是将组合问题中的数字换成括号，并加入判断条件。
**额外约束**：对于括号字符串的任意前缀，右括号的个数不能超过左括号的个数。比如` ())( `是不合法的。

#### 1. 枚举当前位置 左括号 or 右括号 (选左，选右)

本质是「选或不选」的思想，你可以把填左括号视作「选」，填右括号视作「不选」。

递归的过程中，要保证右括号的个数不能超过左括号的个数。
- 如果现在右括号个数等于左括号个数，那么不能填右括号。
- 如果现在右括号个数小于左括号个数，那么可以填右括号。

由于左括号个数始终 ≥ 右括号个数，且至多填 n 个左括号，所以当我们填了 n 个右括号时，也一定填了 n 个左括号，此时填完所有 2n 个括号。

问：什么时候需要写恢复现场，什么时候不需要写？

答：下面代码中，如果初始化 path 为空列表，就需要写恢复现场。本题由于所有括号长度都是固定的 2n，我们可以创建一个长为 2n 的 path 列表，在递归时直接写入字符（而不是插入字符），这样做无需写恢复现场。

作者：灵茶山艾府
链接：https://leetcode.cn/problems/generate-parentheses/solutions/2071015/hui-su-bu-hui-xie-tao-lu-zai-ci-pythonja-wcdw/

```python
from typing import List
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        ans = []
        path = [''] * (n*2)

        #目前填了 left 个左括号，right 个右括号
        def dfs(left: int, right: int):
            if right == n: # 由于始终 left >= right，)填完 ，(填完
                ans.append(''.join(path))
                return

            if left < n: # 可填左括号
                path[left + right] = '(' # 直接覆盖
                dfs(left + 1, right)

            if right < left: # 可以填右括号
                path[left + right] = ')' # 直接覆盖
                dfs(left, right + 1)

        dfs(0, 0)
        return ans
```

#### 2. 枚举下一个左括号位置

用「枚举选哪个」的思路。

在从左往右填的过程中，要时刻保证右括号的个数不能超过左括号的个数。

如果前面填了 5 个左括号，2 个右括号，那么还能填几个右括号？
- 至多填 5−2=3 个。
- 所以枚举（在填下一个左括号之前）填入了 0,1,2,3 个右括号，这样就能得到下一个左括号的位置。

为了方便，代码直接用 balance 表示左右括号之差。这样我们枚举的范围就是 [0,balance]。


```python
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        ans = []
        path = []  # 记录左括号的下标

        # 目前填了 i 个括号
        # balance = 这 i 个括号中的左括号个数 - 右括号个数
        def dfs(i: int, balance: int):
            if len(path) == n:
                s = [')'] * (n * 2)
                for j in path:
                    s[j] = '('
                ans.append(''.join(s))
                return 
            
            # 当前可填 balance 个 )
            # 枚举填 right=0,1,2,...,balance 个右括号
            for right in range(balance + 1):
                # 先填 right 个右括号，然后填 1 个左括号，记录左括号的下标 i+right
                path.append(i + right)
                dfs(i + right +1, balance - right + 1)
                path.pop() # 恢复现场
            
        dfs(0, 0)
        return ans
```
