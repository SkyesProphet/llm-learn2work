### 78.子集
给你一个整数数组 nums ，数组中的元素 互不相同 。返回该数组所有可能的子集（幂集）。

解集 不能 包含重复的子集。你可以按 任意顺序 返回解集。
每个子集里的元素顺序要和原数组一致

示例 1：

输入：nums = [1,2,3]
输出：[[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]

示例 2：

输入：nums = [0]
输出：[[],[0]]

#### 1. 递归 枚举选哪个 dfs + 撤销选择
枚举子集（答案）的第一个数选谁，第二个数选谁，第三个数选谁，依此类推。

1. dfs 中的 i 表示现在要枚举选 nums[i] 到 nums[n−1] 中的一个数，添加到当前 path 末尾。

如果选 nums[j] 添加到 path 末尾，那么下一个要添加到 path 末尾的数，就要在 nums[j+1] 到 nums[n−1] 中枚举了。

此时不用判断，nums序列是否遍历结束，不需要在回溯中判断 i=n 的边界情况，因为此时不会进入循环。

```python
from typing import List
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        res = [] # 记录结果
        path = [] # 记录当前正在构建的路径

        def dfs(index):
            res.append(path.copy()) # 将当前已有路径加入结果中

            for i in range(index, len(nums)):
                path.append(nums[i]) # 选择当前元素构建path
                dfs(i + 1) # 以当前path为前缀，继续构建后续可能的路径，并先保存已有子序列

                # 回退，撤销选择，当前节点继续尝试下一个元素
                path.pop()

        dfs(0) # 从索引0开始构建路径
        return res

# 示例用法
solution = Solution()
nums = [1, 2, 3]
print(solution.subsets(nums))
```

#### 2. 回溯 dfs深度遍历，此节点 选或不选
对于输入的 nums，考虑每个 nums[i] 是选还是不选，由此组合出 2^n 个不同的子集。

dfs 中的 i 表示当前考虑到 nums[i] 选或不选。
1. 递归终止条件：i == n，说明已经考虑完 nums 中的所有元素。
2. 递归调用：
    - 不考虑 nums[i]，继续递归调用 dfs(i + 1)。
    - 考虑 nums[i]，将 nums[i] 添加到 path 中，并继续递归调用 dfs(i + 1)。
3. 回溯：从 path 中移除 nums[i]。

```python
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        res = []
        path = []

        def dfs(index):
            if index == len(nums): # 完整遍历了一遍选择路径
                res.append(path.copy()) # 将当前已选择好的路径加入结果中
                return
            
            # 不选择当前元素，继续选择构建后续路径
            dfs(index + 1)
            # 选择当前元素，加入路径中,继续构建后续路径
            path.append(nums[index])
            dfs(index + 1)
            path.pop() # 恢复现场，撤销当前元素的选择，返回上一层
    
        dfs(0)
        return res
```
