### 77.组合
给定两个整数 n 和 k，返回范围 [1, n] 中所有可能的 k 个数的组合。

你可以按 任何顺序 返回答案。

示例 1：

输入：n = 4, k = 2

输出：
[
  [2,4],
  [3,4],
  [2,3],
  [1,2],
  [1,3],
  [1,4],
]

#### 1. 选哪一个
选哪一个的技巧在： 
1. 从当前位置往后顺序遍历选择，深入下一层，
2. 回退恢复当前位置。
3. 不回头，以避免组合重复。

```python
from typing import List
class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
        # 下一个选什么
        ans = []
        path = []

        # 枚举选哪一个，选的数范围在[1, n]倒着枚举，不回头
        # 选 1 到 i 中选一个，加到path末尾
        def dfs(i: int) ->None:
            d = k - len(path) # 还要选 d个数
            if d == 0:
                ans.append(path.copy())
                return
            
            # 枚举的数不能太小，否则后面没有数可以选
            for j in range(i, d -1, -1):
                path.append(j)
                dfs(j -1) # 
                path.pop()
        dfs(n)
        return ans
```

#### 2. 选或不选
选或不选的技巧在：
1. 结束条件为遍历完 序列，或提前达到target。 添加当前序列，return
2. 当前位置选与不选， 两种情况依次写，都会涉及到：
   1. 不选，进入下一位置的 选与不选
   2. 选择，判断下一状态的 选与不选，并恢复状态

```python
class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
        # 选与不选
        ans = []
        path = []

        # 当前位置选与不选，判断条件为余下需要选的数
        def dfs(i :int) -> None:
            # 判断是否满足条件
            d = k - len(path) # 还要选 d 个数
            if d == 0:  # 达成条件
                ans.append(path.copy())
                return

            # 从后往前选
            # 不选 i
            #由于 i 范围为1-n, 从后遍历,i即为剩余 可选数 的数量
            if i > d:  # i <= d时，余下可选数量不够，为达成条件必须选
                dfs(i - 1)

            # 选 i
            path.append(i)
            dfs(i - 1)
            path.pop()

        dfs(n)
        return ans 
```
