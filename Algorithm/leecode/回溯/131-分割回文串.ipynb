### 131.分割回文串
给你一个字符串 s，请你将 s 分割成一些 子串，使每个子串都是 回文串 。返回 s 所有可能的分割方案。

示例 1：

输入：s = "aab"
输出：[["a","a","b"],["aa","b"]]

示例 2：

输入：s = "a"
输出：[["a"]]

提示：
1 <= s.length <= 16
s 仅由小写英文字母组成




#### 1. 回溯套路 （“，”逗号选或不选）——输入视角
分割子串，相当于对字符串进行切割，切割的点是回文串的边界点。

变化为 每两个字符之间，切割 或 不切割。

理解成：是否要在 i 和 i+1 处分割，把 s[i] 作为当前子串的最后一个字符，把 s[i+1] 作为下一个子串的第一个字符。

注意 `s[n−1]` 一定是最后一个字符，所以在 i=n−1 的时候一定要分割。

**答疑**：
问：代码遇到回文串就加入 path，如何保证这种分割方案一定合法？

答：如果最后一个字符串不是回文串，我们不会递归到 i=n 的边界，不会把不合法的分割加入答案。

**判断回文串**：
    1.：双指针 向中间遍历判断。
    2. reverse, 对比是否相等？： t = s[i:j+1], 判断 if t == t[::-1]:

作者：灵茶山艾府
链接：https://leetcode.cn/problems/palindrome-partitioning/solutions/2059414/hui-su-bu-hui-xie-tao-lu-zai-ci-pythonja-fues/


```python
from typing import List
class Solution:
    def partition(self, s: str) -> List[List[str]]:
        # 当前位置切或不切
        ans = []
        path = []
        n = len(s)

        # 考虑 i 后面的逗号怎么选
        # start 表示当前这段回文子串的开始位置
        def dfs(i: int, start: int):
            if i == n: # 遍历完所有位置，最后一定切割
                ans.append(path[:])
                return 
            
            # 不切，不选 i 和 i+1 之间的逗号
            if i < n - 1: # i=n-1 时只能分割,最后一个字符了不能不切了
                dfs(i + 1, start) # 开始位置不变，判断后续切不切
            # 切
            if s[start:i + 1] == s[start:i + 1][::-1]:
                path.append(s[start:i + 1]) # 切割当前子串,左闭右开
                dfs(i + 1, i + 1) # 切了之后，开始位置变为 i + 1
                path.pop()
        
        dfs(0, 0)
        return ans
    
s = ["aab"]
print(Solution().partition(s[0]))
```

#### 2. 枚举子串结束的位置——答案视角
枚举回文子串结束的位置，即 枚举 切割 的位置：
- 枚举第一个 切割 位置，或没有
- 枚举第二个 切割 位置，或没有

```python
from typing import List
class Solution:
    def partition(self, s: str) -> List[List[str]]:
        # 依次枚举切割的位置，遍历完所有位置必须切割一次
        ans = []
        path = [] # 会遍历完所有字符，切割字符串保证每个字符都有
        n =len(s)

        def dfs(i :int) -> None:
            # 递归边界,切割完所有字符
            if i == n:
                ans.append(path.copy())
                return
            # 从当前位置开始的字符串，依次枚举切割的位置
            for j in range(i, n):
                # 判断切割的字符串是否是回文串
                # if self.isPalindrome(s, i, j):
                #     path.append(s[i:j+1])
                #     dfs(j+1)
                #     path.pop()
                t = s[i:j+1] # 左闭右开
                if t == t[::-1]:
                    path.append(t)
                    dfs(j+1) # j切割后，继续切割 j+1位置开始的字符串
                    path.pop() # 增添型路径记录，恢复现场

        dfs(0)
        return ans

s = ["aab"]
print(Solution().partition(s[0]))
```
