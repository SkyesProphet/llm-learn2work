### 46.全排列
给定一个不含重复数字的数组 nums ，返回其 所有可能的全排列 。你可以 按任意顺序 返回答案。

对于一个长度为 n 的数组（假设元素互不重复），其排列方案数共有：
    `n×(n−1)×(n−2)…×2×1`


#### 回溯法——（类深度优先搜索）


```python
# 回溯，直接在原数组上修改
from typing import List
class Solution:
    def permute(self, nums: List[int]) -> List[List[int]]:
        res = [] # 记录所有序列

        def dfs(depth): # 深度优先搜索
            # depth为当前确定的深度数
            if depth == len(nums) -1: #已遍历到最深
                res.append(list(nums)) # 添加排列方案，列表化
                return

            for i in range(depth, len(nums)): # 对于后续剩余位置数
                nums[i], nums[depth] = nums[depth], nums[i] # 交换当前深度数
                dfs(depth +1) # 继续遍历下一个深度
                nums[i], nums[depth] = nums[depth], nums[i]

        dfs(0)
        return res
```

```python
# 同样回溯，但创建新序列，并记录状态
class Solution:
    def permute(self, nums: List[int]) -> List[List[int]]:

        n = len(nums) # 数组长度，即深度
        path = [0] * n # 所有排列的长度都是 n
        on_path = [False] * n
        res = [] # 记录所有序列

        def dfs(x: int) -> None:
            if x == n:
                res.append(path[:]) # or path,copy()
                return 
            
            for i, on in enumerate(on_path): # 位置下标和状态
                if not on: # on 为false,表示数字没有被选中
                    path[x] = nums[i] #从没有选的数字中选一个
                    on_path[i] = True # 已选上
                    dfs(x + 1) # 下一层深度
                    on_path[i] = False # 恢复现场
                    # 注意 path 无需恢复现场，因为排列长度固定，直接覆盖就行
        dfs(0)
        return res            
```
