### 142.环形链表Ⅱ
给定一个链表的头节点  head ，返回链表开始**入环的第一个节点**。 如果链表无环，则返回 null。

如果链表中有某个节点，可以通过连续跟踪 next 指针再次到达，则链表中存在环。 为了表示给定链表中的环，评测系统内部使用整数 pos 来表示链表尾连接到链表中的位置（索引从 0 开始）。如果 pos 是 -1，则在该链表中没有环。注意：pos 不作为参数进行传递，仅仅是为了标识链表的实际情况。

不允许修改 链表。

#### 快慢指针
双指针：寻找距离尾部第 K 个节点、寻找环入口、寻找公共尾部入口等

本题的求解过程中，双指针会产生两次“相遇”。

快慢指针的第一次相遇(一定在 环内)：
分析：
**第一次相遇**
1. 设链表共有 a+b 个节点，其中 链表头部到链表入口 有 a 个节点（不计链表入口节点）， 链表环 有 b 个节点。
2. fast 走的步数是 slow 步数的 2 倍，即 `f = 2s`；
3. fast 比 slow 多走了 n 个环的长度，即 `f = s+nb`。(重合时 fast 比 slow 多走 环的长度整数倍) fast 追上 slow， 必定夺走 n 圈。
4. 两式相减得到 `f = 2nb，s = nb`，即 fast 和 slow 指针分别走了 `2n，n` 个环的周长。

**第二次相遇**
如果让指针从链表头部一直向前走并统计步数k，那么所有 走到链表入口节点时的步数 是：`k=a+nb` ，即先走 a 步到入口节点，之后每绕 1 圈环（ b 步）都会再次到入口节点。
1. 目前 slow 指针走了 nb 步。让 slow 再走 a 步，即 slow 走到入口节点。
2. 令 fast 重新指向链表头部节点。此时 f=0，s=nb 。
3. 当 fast 指针走到 `f = a` 步时，slow 指针走到 `s = a+nb` 步。此时两指针重合，并同时指向链表环入口，返回 slow 指向的节点即可。


```python
from typing import Optional
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

class Solution:
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        fast = slow = head

        while fast and fast.next:
            fast = fast.next.next #+2
            slow = slow.next # +1

            if slow == fast: # 相遇点
                index = head # 从头 和相遇点 同时出发，直到再次相遇
                while index != fast: # 再次相遇点 就是入口点
                    index = index.next
                    fast = fast.next
                return index # 在入口处结束循环

        return None # no cycle
```
