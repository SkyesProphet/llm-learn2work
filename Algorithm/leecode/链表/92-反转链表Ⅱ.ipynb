### 92. 反转链表 II
给你单链表的头指针 head 和两个整数 left 和 right ，其中 left <= right 。请你反转从位置 left 到位置 right 的链表节点，返回 反转后的链表 。

示例 1：
输入：head = [1,2,3,4,5], left = 2, right = 4

输出：[1,4,3,2,5]

示例 2：
输入：head = [5], left = 1, right = 1

输出：[5]


#### 1. 区间翻转链表 - 双指针
先定位到 left 指针起始位置，left 前节点作为 伪头节点，right 后一节点为 后续链表头。

循环迭代`right - left +1` 次依次移动翻转。将翻转后的链表相连。


```python
from typing import List, Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def reverseBetween(self, head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:
        if not head or left == right: 
            return head

        # 先定位到 left节点 前，依次迭代right -left + 1 次进行翻转
        # left , right 序号 从 1 开始
        p0 = dummy = ListNode(next = head)
        for _ in range(left - 1):
            p0 = p0.next

        # 此时 p0在 翻转子链表 前
        pre, cur = None, p0.next # 前序节点，翻转头节点
        for _ in range(right - left + 1): # 需翻转 right - left + 1次
            tmp = cur.next
            cur.next = pre
            pre = cur
            cur = tmp
        
        # 翻转后 pre为新子链头节点, 新子链尾节点为 p0.next
        # cur 为 后续子链头
        dum_tail = p0.next
        dum_tail.next = cur # 连接后续子链
        p0.next = pre
        
        return dummy.next 
```
