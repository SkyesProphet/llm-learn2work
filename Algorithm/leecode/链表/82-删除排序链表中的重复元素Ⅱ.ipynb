### 82. 删除排序链表中的重复元素 II
给定一个已排序的链表的头 head ， 删除原始链表中所有重复数字的节点，只留下不同的数字 。返回 已排序的链表 。

示例1：

输入：head = [1,2,3,3,4,4,5]
输出：[1,2,5]

示例2：

输入：head = [1,1,1,2,3]
输出：[2,3]

题目数据保证链表已经按升序 排列。

#### 去重复元素
题目要求删除所有重复元素，即有重复的元素全删除，不保留。

由于题目明确当前链表 已升序排序，故重复元素一定相邻。

因此，只需要判断**连续两个节点**是否相等，相等则暂存此值，遍历出所有当前重复元素，并删除。

1. 使用迭代，创建一个哑节点。 cur = dummy = ListNode(next = head)开始
2. 当前节点作为 两判断节点的起始。
3. 判断 接下来连续两节点的值是否相等，相等则暂存此值，遍历出所有当前重复元素，并删除。

```python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def deleteDuplicates(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 迭代，由于链表已排序，每两个节点判断是否为重复元素
        cur = dummy = ListNode(next=head)

        while cur.next and cur.next.next: # 可比较两个节点
            val = cur.next.val
            if cur.next.next.val == val: # 存在重复元素
                while cur.next and cur.next.val == val: # 遍历寻找重复元素
                    cur.next = cur.next.next  # 删除下一个重复节点

            else:
                cur = cur.next
        return dummy.next
```
