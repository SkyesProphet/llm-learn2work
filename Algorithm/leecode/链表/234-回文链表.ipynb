### 234. 回文链表
给你一个单链表的头节点 head ，请你判断该链表是否为回文链表。如果是，返回 true ；否则，返回 false 。

示例 1：
输入：head = [1,2,2,1]
输出：true

示例 2：
输入：head = [1,2]
输出：false

提示：
链表中节点数目在范围[1, 10^5] 内
0 <= Node.val <= 9

进阶：你能否用 O(n) 时间复杂度和 O(1) 空间复杂度解决此题？

#### 1. 暴力数组法
直接遍历链表保存伪数组，然后判断数组是否为回文数组

```python
# 暴力数组法
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        arr = []
        tmp = head
        while tmp:
            arr.append(tmp.val)
            tmp = tmp.next
        return arr == arr[::-1] # 判断数组是否是回文数组
    
```

#### 2.快慢指针法——找中点

判断链表回文串：1. 找链表中点； 2. 反转后半部分链表； 3. 比较前后两半链表。
1. 找中点：快慢指针法
   1.如果链表有 奇数个节点，找正中间节点，为第 (n + 1) / 2 个节点。
   2.如果链表有偶数个节点，找正中间右侧节点，为第 n / 2 个节点。 
2. 反转后半链表：迭代法
3. 同时遍历 head 和 head 2 开始比较，如果相等则继续比较，否则返回false。 

```python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        # 快慢指针
        fast = slow = head
        # 快慢指针找中点
        while fast and fast.next: # 此时slow在中点
            fast = fast.next.next
            slow = slow.next

        pre = None
        # 反转后半部分链表
        while slow:
            tmp = slow.next # 暂存slow的下一个节点
            slow.next = pre # 反转
            pre = slow # pre向前移动
            slow = tmp # slow向前移动

        # pre是后半部分链表的头节点
        while pre: # 由于后半节点一定 小于等于前半，遍历后半
            if pre.val != head.val:
                return False
            pre = pre.next
            head = head.next
        return True
```

```python
# 各功能模块化
class Solution:
    def middleNode(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
        return slow

    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        pre = None
        cur = head
        while cur:
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt
        return pre

    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        if not head or not head.next:
            return True
        
        mid = self.middleNode(head)
        right = self.reverseList(mid)
        while right:
            if right.val != head.val:
                return False
            right = right.next
            head = head.next
        return True
        
```
