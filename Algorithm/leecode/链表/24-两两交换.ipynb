### 24. 两两交换链表中的节点
给你一个链表，两两交换其中相邻的节点，并返回交换后链表的头节点。你必须在不修改节点内部的值的情况下完成本题（即，只能进行节点交换）。

示例 1：
输入：head = [1,2,3,4]
输出：[2,1,4,3]

示例 2：
输入：head = []
输出：[]

示例 3：
输入：head = [1]
输出：[1]

提示：
链表中节点的数目在范围 [0, 100] 内
`0 <= Node.val <= 100`

#### 1. 迭代
每两个节点一组， 翻转节点，并连接 后续节点。
步骤：
1. 创建哑节点，方便返回结果
2. 创建两个指针，分别指向当前节点和当前节点的下一个节点。
3. 迭代，每两个节点一组，翻转节点，并连接后续节点。

```python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        pre = dummy = ListNode(next=head) # 创建虚拟头
        cur = head

        while cur and cur.next: # 至少两个节点
            nxt = cur.next # 临时保存当前节点的下一个节点
            tmp = nxt.next # 下一组子链表头节点

            # 当前节点反转
            pre.next = nxt
            nxt.next = cur
            cur.next = tmp 

            # 更新指针
            pre = cur
            cur = tmp # 下一轮交换的节点

        return dummy.next
```

#### 2. 递归
和方法1 类似。 将后续节点处理 视为 子问题（递归问题）。
递归步骤： 直接将 `swapPairs` 函数当作 递归函数。
1. 递归边界：如果 head 或者 head.next 为空，说明剩余节点不足两个，无需交换，返回 head。
2. 先 处理子问题，即先 递归，递归返回值 为已交换好的子链。
3. 交换当前两节点，返回交换后的头节点。

```python
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 递归终止条件：当前节点或下一个节点为空
        if not head or not head.next: # 没有两个节点，递归边界 
            return head
        
        cur = head
        nxt = head.next # 临时保存当前节点的下一个节点
        tmp = nxt.next # 下一组子链表头节点

        # 先递归，确保当前交换后的子链是 处理好的
        cur.next = self.swapPairs(tmp) # 递归交换下一组节点
        nxt.next = cur # 当前节点反转
        return nxt # 返回新的头节点
```
