### 206. 反转链表
给你单链表的头节点 head ，请你反转链表，并返回反转后的链表。

输入：head = [1,2,3,4,5]

输出：[5,4,3,2,1]

输入：head = [1,2]

输出：[2,1]

输入：head = []

输出：[]

#### 1. 递归法
将 链表当成 两部分 pre 和 cur, 进行简单的 cur->pre 翻转操作。
由于使用递归，实则从后往前翻转。但需要将 **最后一个节点**的指针保存起来，作为返回值传递。
- 递归终止条件：cur == null，则返回尾节点 pre, 返回值记作 res 传递。
- 修改当前节点 cur 的指针方向：cur.next = pre
- 返回最终链表头节点 res

```python
from typing import Optional
class ListNode:
    def __init__(self, val = 0, next = None):
        self.val = val
        self.next = next

class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 当成两个部分进行递归
        pre, cur = None, head
        def recur(cur, pre):
            if not cur: # 递归结束条件
                return pre # 返回的其实是尾节点
            # 递归调用
            res = recur(cur.next, cur)
            cur.next = pre # 反转
            return res # 返回头节点

        return recur(cur, pre)
```

#### 2. 双指针迭代法
遍历链表，并在访问各节点时修改 next 引用指向，算法流程见注释。

```python
class Solution:
    def reverseList(self, head: ListNode) -> ListNode:
        pre, cur = None, head
        while cur:
            tmp = cur.next # 暂存下一个节点
            cur.next = pre # 反转
            pre = cur # pre 向前移动
            cur = tmp # cur 移动到下一个节点
        return pre # 最后 pre 就是新的头节点
```
