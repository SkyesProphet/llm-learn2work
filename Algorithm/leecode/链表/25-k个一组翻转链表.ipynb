### 25. K个一组翻转链表
给你链表的头节点 head ，每 k 个节点一组进行翻转，请你返回修改后的链表。

k 是一个正整数，它的值小于或等于链表的长度。如果节点总数不是 k 的整数倍，那么请将最后剩余的节点保持原有顺序。

你不能只是单纯的改变节点内部的值，而是需要实际进行节点交换。

示例 1：

输入：head = [1,2,3,4,5], k = 2

输出：[2,1,4,3,5]

示例 2：

输入：head = [1,2,3,4,5], k = 3

输出：[3,2,1,4,5]

#### 1. 递归法 + 迭代
递归翻转 = 翻转每个 子链表 前 k 个节点
1. 判断剩余元素是否达到 k 个：若没达到，就直接返回结果；若达到，进行翻转
2. 翻转 k 个节点（三指针，迭代）

```python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        # 递归法
        # 翻转链表前k个节点，翻转每个子链表前K个节点

        # 思路：
        # 先检查当前链表是否可组成一组K个，可以则处理当前组的内部翻转
        curr = head
        count = 0
        while curr and count < k:
            curr = curr.next
            count += 1
        
        if count < k:
            return head # 若不足k个不处理直接返回 

        # 翻转链表，三指针法（prev, curr, next）
        prev = None
        curr = head
        for _ in range(k): # 逐步翻转链表箭头
            next_temp = curr.next
            curr.next =prev
            prev = curr
            curr = next_temp

        # 剩下的交给子任务继续分组处理
        # 翻转后最初的头变尾 所以 head.next
        head.next = self.reverseKGroup(curr, k)

        # - prev 是当前 k 个节点翻转后的头
        return prev
```

#### 2. 迭代法：k 个 分组翻转
依次 k 个一组翻转

```python
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None
class Solution:
    def reverseKGroup(self, head: ListNode, k: int) -> ListNode:
        # k 个节点翻转链表
        def reverse(head, k):
            pre, cur = None, head
            for _ in range(k):
                next = cur.next
                cur.next = pre
                pre = cur
                cur = next
            return pre, cur # 此时 pre为翻转后 头节点，cur为下一个子链表头节点
        
        # 迭代法
        # 统计节点个数
        n = 0
        cur = head
        while cur:
            cur =cur.next
            n += 1

        pre = dummy = ListNode(0)
        dummy.next = head
        # 每次翻转 k 个节点
        while n >= k:
            n -= k # 更新剩余节点个数
            re_head, cur = reverse(pre.next, k) # 翻转当前 k 个节点
            pre_tail = pre.next # 翻转前的头节点变成翻转后的尾节点
            pre_tail.next = cur # 连接翻转后的尾节点和下一个子链表头节点
            pre.next = re_head # 连接前一个子链表尾节点和翻转后的头节点
            pre = pre_tail # 更新 pre 指针到当前子链表尾节点，为下一次翻转做准备

        return dummy.next
    
l = [1,2,3,4,5]
k = 2
# 构建链表
head = ListNode(l[0])
cur = head
for i in range(1, len(l)):
    cur.next = ListNode(l[i])
    cur = cur.next

s = Solution()
new_head = s.reverseKGroup(head, k)
# 输出翻转后的链表，链表转为数组形式输出
result = []
cur = new_head
while cur:
    result.append(cur.val)
    cur = cur.next
print(result)
```

#### 3. 暴力生成
强制将 链表 转换成 数组，然后使用 暴力生成 新链表。

```python
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        #强制转换 转化为数值，数组变换后再生成新链表
        lst = []
        while head:
            lst.append(head.val)
            head = head.next    # 转换为数组
        for i in range(0, len(lst)-k+1, k): # range(start, stop, step) 起始位，截至位，步幅
            lst[i:i+k] = lst[i:i+k][::-1]  # 列表切片翻转[::-1]
 
        head_new = ListNode(lst[0])
        curr = head_new
        for v in lst[1:]:
            curr.next = ListNode(v)
            curr = curr.next
        return head_new
```
