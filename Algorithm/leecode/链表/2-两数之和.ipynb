### 2. 两数之和

给你两个 非空 的链表，表示两个非负的整数。它们每位数字都是按照 逆序 的方式存储的，并且每个节点只能存储 一位 数字。

请你将两个数相加，并以相同形式返回一个表示和的链表。

你可以假设除了数字 0 之外，这两个数都不会以 0 开头。

示例 1：

输入：l1 = [2,4,3], l2 = [5,6,4]
输出：[7,0,8]
解释：342 + 465 = 807.

示例 2：

输入：l1 = [0], l2 = [0]
输出：[0]

示例 3：

输入：l1 = [9,9,9,9,9,9,9], l2 = [9,9,9,9]
输出：[8,9,9,9,0,0,0,1]

#### 1. 递归法
由于链表倒着 存储 数值的每个 位数；且 两数相加时 从低位开始相加，所以 从头遍历链表，即从低位开始相加。

从 头开始 两链表 节点数两两相加，后续子链表 内容 可视为 和 链表相加 相似的**子问题**。故可用**递归**。

递归步骤：
1. 每次将 两节点值 `l1.val + l2.val + 进位` 相加，除以 10 取商，即 **进位**，**余数** 为 当前位保存的 位数。
2. 终止条件：
   - 递归边界：两个链表都为空，且 进位为 0，返回 None
   - 链表不为空，返回 进位 + 当前位保存的 位数

```python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode], carry = 0) -> Optional[ListNode]:
        # 递归
        # l1 和 l2 为当前遍历的节点，carry 为进位
        # 递归结束条件：l1 和 l2 都为 None，即两个链表都遍历完
        if not l1 and not l2 and carry == 0: 
            return None

        s = carry
        if l1: 
            s += l1.val # 累加节点和进位
            l1 = l1.next
        if l2: 
            s += l2.val
            l2 = l2.next
        
        # 创建当前位置的节点
        # s // 10 为商，s % 10 为余数
        # 节点的 下一个节点为递归调用的结果
        return ListNode(s % 10, self.addTwoNumbers(l1, l2, s // 10))
```

#### 2. 迭代法
迭代的思路是，初始化答案为一个「空链表」，每次循环，向该链表末尾添加一个节点（保存一个数位）。

迭代步骤：
1. 创建一个空链表，哨兵节点作为 伪节点头，用于保存结果。
2. 循环即遍历链表 `l1` 和 `l2`，计算 ` l1.val + l2.val + carry `。除以 10，得到商 `carry`，得到余数 `val`。
3. 创建一个节点，保存余数 `val`，并添加到结果链表的末尾。

```python
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        # 迭代
        cur = dummy = ListNode() # 哑节点，方便返回结果
        carry = 0 # 进位

        while l1 or l2 or carry: # 有节点或者进位时继续迭代
            if l1: # 节点不为空时，取节点值
                carry += l1.val
                l1 = l1.next
            if l2: # 节点不为空时，取节点值
                carry += l2.val
                l2 = l2.next
            cur.next = ListNode(carry % 10) # 创建节点，并保存当前节点
            carry //= 10 # 更新进位
            cur = cur.next # 迭代
        return dummy.next # 返回结果
```
