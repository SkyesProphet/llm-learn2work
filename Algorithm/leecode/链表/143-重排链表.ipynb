### 143.重排链表
给定一个单链表 L 的头节点 head ，单链表 L 表示为：
`L0-L1-L2 ... L(n-1) - L(n)`

重排后变为：
`L0-L(n) - L1 - L(n-1) - L2 ...`

一头一尾交互排序。

不能只是单纯的改变节点内部的值，而是需要实际的进行节点交换。

示例1.：

输入：head = [1,2,3,4]
输出：[1,4,2,3]

示例2：

输入：head = [1,2,3,4,5]
输出：[1,5,2,4,3]

#### 解法：交错合并
题目要求可视为：
1. 将链表对半分
2. 反转 后半链表
3. 将 前半链表 与 后半链表 **交错合并**
   
##### 相关题目
- **876.链表的中间节点**
- **206.反转链表**
    
**实现步骤**：
1. 链表对半，获得 head 链表 和 head2 链表
2. 提前记录head 和 head2 的下一个：` nxt = head.next, nxt2 = head2.next`
3. `head.next = head2, head2.next = head.next` # 实现一组交错合并 
4. 继续迭代合并下一组。


```python
from typing import List, Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    # 1. 找到链表中间节点
    def middleNode(self, head: Optional[ListNode]):
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
        return slow

    # 2. 反转链表
    def reverseList(self, head: Optional[ListNode]):
        prev = None
        curr = head
        while curr:
            next_temp = curr.next
            curr.next = prev
            prev = curr
            curr = next_temp
        return prev

    # 3. 交错合并链表
    def reorderList(self, head: Optional[ListNode]) -> None:
        # 找到链表中间节点
        mid = self.middleNode(head)
        # 反转后半部分链表
        head2 = self.reverseList(mid)
        # 交错合并链表
        while head2.next:
            next1 = head.next # 保存原链表的下一个节点
            next2 = head2.next # 保存后半链表的下一个节点
            
            # 交错合并
            head.next = head2 
            head2.next = next1
            
            # 移动至下一组节点
            head = next1  
            head2 = next2
        # 在原链表上修改，无需返回值
```
