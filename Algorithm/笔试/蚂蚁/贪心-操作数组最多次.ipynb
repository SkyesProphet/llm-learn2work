### 2026.3.19 操作数组最多次(开发岗)
**题目描述**
给定一个长度为n的数组 {a1, a2,..., an} 。你可以进行若干次操作，每次操作从下列两种中任选其一：
- 选择一个元素将其删除，剩余元素按照原顺序拼接；
- 选择一个元素将其增加1或者减少1。
要求：每次操作完成后，数组所有元素的和必须是2的倍数（即为偶数）。请你输出最多可以进行多少次操作。

**输入描述**：
输入测试例的描述：
每个测试文件均包含多组测试数据。第一行输入一个整数t（1≤t≤ 10^4）表示数据组数，每组测试数据描述如下：
- 第一行输入一个整数n（1≤n≤2×10^5）；
- 第二行输入n个整数 a1， a2，⋯，an（1≤ai ≤10^9）表示数组a。
除此之外，保证单个测试文件的 n之和不超过2×10^5。

示例1：
输入：
```text
2
5
1 2 3 4 5
4
2 2 1 1 
```
输出：
```text
4
2
```

#### 贪心：要求最多操作
逻辑推导：
设当前数组所有元素的和为 S。题目要求每次操作后，新的和 S′必须是偶数。
##### 分析操作影响：
1. 删除一个元素：S′ = S - x，其中 x 是被删除的元素。
   1. 要求 S - x 为偶数，**即 S 和 x 奇偶性一致**。
   2. 操作：
      1. 当前和 S 是偶数，只能删除偶数。
      2. 当前和 S 是奇数，只能删除奇数。
2. 将一个元素 增加或减少 1：S′ = S + 1 or S - 1：
   1. 要求 S + 1 or S - 1 为偶数，**即 S 必须为奇数**。
   2. 操作：只有当 S 是奇数时，才能进行操作。修改后，数组长度不变，次数 + 1。

##### 总结情况
1. 初始 S 是偶数，只能删除偶数(操作1)。
   1. S 减去一个偶数，结果仍然是偶数。只要数组中**存在偶数**，就只能一直进行 减偶 操作。
   2. 当偶数删光了，剩下的全是奇数。此时和 S 依然是偶数。
   3. **偶数 S 无法进行删除奇数和加减操作**。终止。
   4. **最大操作次数 = 数组中偶数的个数**。
2. 初始 S 是奇数。（仅当数组中奇数个数为奇数个时，才会出现此种情况）两种操作选择:
   1. 删除奇数： S (奇) - 奇 = 偶。操作后**和变偶数**，进入**情况 1**的状态。
      1. 数组 a 删一个奇数( 1 次操作)，数组元素 -1，偶数个数不变。
      2. **最大操作次数 = 1 + 数组中偶数的个数**。
   2. 修改任意数： S (奇) ± 1 = 偶。操作后**和变偶数**，数组长度不变，进入**情况 1**的状态。
      1. 修改一个数（1 次操作）， 数组长度不变。状态变为：和是偶数。
         1. 若 将 数组中 **奇数修改**，偶数个数 + 1。 总数即为：1 + 偶数个数 + 1。
         2. 若 将 数组中 **偶数修改**，余下偶数个数 -1。 总数即为：偶数个数。
      2. 后续 删除所有偶数。
      3. **最大操作次数 = 2 + 数组中偶数的个数**。（修改奇数）
   3. 综上，选择最大操作数，应该选**路径 2**，修改奇数。


````python
# 1. 数组和 sum 为 偶数时，只能进行删除偶数操作 num = count_even。
# 2. 数组和 sum 为 奇数时，奇数 为奇数个，先修改奇数, 能获得最大操作数。 num = 2 + count_even

# 定义状态
# count_even: 统计偶数的个数。
# count_odd: 统计奇数的个数。
# sum: 数组元素的总和。
import sys

# data = sys.stdin.read().splitlines() # 读取每行字符串

# 输入测试
data = ["2", "5", "1 2 3 4 6", "4", "2 2 2 1"]
iterator = iter(data) # 将输入的多行字符串 变为迭代器


t_str = next(iterator) # 迭代器走一步，获得数组数 t
t = int(t_str) #类型转换

for _ in range(t):

    n = int(next(iterator))
    a_str = next(iterator)
    a = list(map(int, a_str.split()))
    count = len(a) # 总个数
    total = sum(a) # 总和
    count_even = sum(1 for x in a if x % 2 == 0)
    count_odd = count - count_even

    if total % 2 == 0: # 若和为偶数
        res = count_even
    else:# 奇数是，必有奇数，且有奇数个
        res = count_even + 2
    print(res)
````
