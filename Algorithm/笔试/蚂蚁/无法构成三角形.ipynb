### 2026.3.19 无法构成三角形
**题目描述**：
给定一个正整数 n（3≤ n ≤ 2×10^5），表示需要构造的数组长度；构造一个长度为 n 的数组
｛a1, a2，…，an｝，使其满足以下条件：
1 ≤ ai ≤ 10^9;

任意选择三个不同下标 i, j, k，设相应的数组元素按非降序排列为 x ≤ y ≤ z，则有 x + y
≤ z，即无法组成三角形。

**输入描述**
在一行上输入一个整数 n（3≤ n ≤ 2×10^5）, 表示数组长度。

**输出描述**
若存在满足条件的数组，则输出一行 n 个整数 a1, a2，…，an；否则，输出 -1。

如果存在多个解决方案，您可以输出任意一个，系统会自动判定是否正确。注意，自测运行功能可能因此返回错误结果，请自行检查答案正确性。

示例：
输入
3
输出
1 2 4

#### 任意两项和小于下一项：斐波那契数列变体

由于 任意下标数 会 非降序排列，使 x + y < z。 构造升序数组，使得 任意两项和小于或等于下一项。
**需要注意输入限制，考虑边界**

斐波那契数列： Fib[i] = Fib[i-1] + Fib[i-2]，满足 x + y = z

**通项规律**：
1. 只要数组是非降序排列，且满足`a[i] = a[i-1] + a[i-2]`（或者更大）;
2. 那么对于任意 `i< j < k`，必然有 `a[i] + a[j] <= a[j] + a[j-1] <= a[j+1] <= a[k]`

特别是对于最小的两个数 a1,a2 和最大的数 ak，只要a1 + a2 ≤ a3，且后续增长足够快，就能满足所有组合。

**数值范围检查！**
1. 输入限制：`3 <= a.length <= 2 *10^5`，`1 <= a[i] <= 10^9`。
2. 斐波那契数列：`Fib[45] ~= 1.13* 10^9 > 10^9`。
3. 给定 n 数组长范围`1 <= n <= 10^5`，当 n > 45 时，斐波那契数列无法满足要求。
4. 由递推下界可知，数组**最小满足情况**就是 **斐波那契数列**。对于 n > 45 的情况，不可能存在满足条件的正整数数组（因为任何满足条件的数组，其第 k 项至少是斐波那契数列的第 k 项）。
5. 根据题目要求，当 n > 45 时，返回 -1。


```python
import sys

n = int(input().strip())
# n = 5

fib = [1, 1]

# 先判断 n >= 45，如果大于45，直接输出-1
if n >= 45:
    print(-1)
else: # n < 45， 构造斐波那契数列
    for i in range(2, n):
        fib.append(fib[i-1] + fib[i-2])
    print(" ".join(map(str, fib[:n])))
```
