### 2026.3.15 搭房子
小明想用木棍搭一个“房子"形状:下部是一个长方形，上部是一个等腰三角形，且两者共用一条边，即长方形的上边同时作为三角形的底边(三角形除底边的两条边要相等)。他手上有一堆木棍(每根木棍的长度为正整数)，每根木棍最多使用一次。

请你判断，是否能从中挑出若干根，恰好拼成这样的"房子"。

**输入描述**：
输入一行一个整数n(6≤ n ≤10^5)，表示木棍数量。
输入第二行包含 n 个整数 a1, a2, ..., an，表示每根木棍的长度。

**输出描述**：
输出一个整数，表示是否能拼出"房子"。如果能，输出 YES；否则，输出 NO。

输入样例：
```text
6
4 4 3 3 5 5
```
输出样例：
```text
YES
```
或者：
```text
6
4 4 3 3 2 1
```
```text
NO
```

题目分析：
1. 判断 输入的数 n > 6。 等腰三角形和长方形房子，需三对等长边。
2. 统计频率：使用哈希表（字典）或数组统计每种长度木棍的数量。
3. 共用边 B：
   - 所有出现过的长度 b，若 dic[b] >= 2，可将其作为共用边。
   - dic[b] -= 2
4. 要求三角边 2*K > B，即 K > B/2。且 dic[K] >= 2。
5. 长方形宽随意，dic[c] >= 2，即可作为长方形的宽。

分析判断：
1. 当 2 K < B 时，由于仍存在 2 B > K 的情况，故 交换位置，仍然能构成房子。
2. 极简判定，存在 三对 等长边， 输出 YES。否则输出 NO。
   1. 统计每种木棍数量 Counter 
   2. 循环遍历字典判断三次，每次取出两条等长边，dic[i] -=2
   3. 若 循环中 无法达成三对等长边，则输出 NO。否则输出 YES。

````python
import sys
from collections import Counter

input = sys.stdin.readline

# data = sys.stdin.read().splitlines() # 按行全部读取字符串
data = ['6', '4 4 3 3 5 5']
iterator = iter(data)
n = int(next(iterator)) # 输入木棍数量
a = list(map(int,next(iterator).split()))


# n = int(input()) # 输入木棍数量
# a = list(map(int,input().split())) # 输入木棍长度列表

dic = Counter(a) # 统计每种长度木棍的数量
# 计算总共有多少对相同的木棍
total = 0
for num in dic.values():
    if num >= 2:
        total += num // 2

print("YES" if total >= 3 else "NO")
````
