### 2026.3.25 圣诞老人分糖果
圣诞老人有 n 种糖果，第 i 种糖果有 ci 个。他要把这些糖果分给 k 个小朋友。分配规则如下：
1. 每一种糖果必须全部分完。
2. 对于任意一种糖果，任意两个小朋友所分得的该种糖果的数量之差不超过 1。
   
在所有合法分配中，任选一个小朋友，统计其最终得到的糖果总数；求该数值在所有合法分配方案中可能达到的最小值与最大值。

换句话说，求在所有合法分配方案中，任意一个小朋友所能获得的糖果总数最小值与最大值。

**输入描述**
每个测试文件均包含多组测试数据。第一行输入一个整数 T（1 ≤ T ≤ 2*10^5），表示测试数据的组数。

1. 第一行输入两个整数 n, k （1 ≤ n ≤ 2*10^5, 1 ≤ k ≤ 10^9）
2. 第二行输入 n 个整数 c1, c2, ..., cn （1 ≤ ci ≤ 10^9）
3. 除此之外，保证单个测试文件的 n 之和不超过 2*10^5

**输出描述**
对于每组测试数据，输出两个整数：在所有合法分配中，单个小朋友可能得到的糖果总数的最小值与最大值。

输入样例：
```text
3
3 2
5 2 7
4 3
0 0 0 0
2 5
10 1
```
输出样例：
```text
6 8
0 0
2 3
```

#### 1. 贪心解法
均分每种糖果，每种糖果余下 ci // k = x。第 i 种糖果的余数 x 会有 x 个小朋友多拿 1 个。分配人数可以随意。

贪心策略下，会有小朋友每种 i 糖果的余数都分到 + 1，也有会小朋友一颗多余糖果没有，达到最大最小差异。

1. 对于第 i 种糖果，设 `base_i = ci // k`，余数 `xi = ci % k`。
2. 每个小朋友至少从第 i 种糖果拿到 base_i 个，其中 xi 个小朋友多拿 1 个。
3. 最大值为：`sum(base_i) + cnt`, cnt 为 xi 的个数。
4. 最小值：sum(base_i) + max(0, sum(xi) - (k - 1)* n)
   - 尽量不让某小朋友拿到 多 1。由于每个小朋友最多多拿 n*1 个, 设其余 k-1 个小朋友最多承担 `(k -1) * n`个，当`sum(xi) <= (k -1) * n`, 目标小朋友可以不拿任何"多1"；否则至少拿 `sum(xi) - (k -1) * n`个"多1"。
   - 简化理解：必定有 xi < k, 故每种糖果 i 至少有 1 个小朋友没有多拿。
   - `min_ans = sum(base_i)`  


````python
import sys

# data = sys.stdin.read().splitlines() # 按行全部读取字符串
data = ['3', '3 2', '5 2 7', '4 3', '0 0 0 0', '2 5', '10 1'] # 测试样例
iterator = iter(data)
t = int(next(iterator))  # 测试用例数
ans = []
for _ in range(t):
    n, k = map(int, next(iterator).split()) # 糖果种类数，小朋友数
    c = list(map(int, next(iterator).split())) # 每种糖果的数量

    min_ans, max_ans = 0, 0

    for num in c: # 遍历每种糖果的数量
        a = num // k # 每种糖果平均分给小朋友的数量
        x = num % k # 剩余的糖果数量
        min_ans += a
        max_ans += a
        if x > 0:
            max_ans += 1
    # print(min_ans, max_ans)

    ans.append(f"{min_ans} {max_ans}")

print("\n".join(ans))
````
